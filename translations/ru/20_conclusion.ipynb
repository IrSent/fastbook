У вас есть много, что сказать. Возможно, вы проводили эксперименты с набором данных, который другие люди не рассматривали таким образом. Расскажите об этом миру! Или, возможно, вы думали о том, чтобы попробовать некоторые идеи, которые пришли вам в голову во время чтения – сейчас самое подходящее время, чтобы превратить эти идеи в код.

Если вы хотите поделиться своими идеями, один из несложных способов – это [форумы fast.ai](https://forums.fast.ai/). Там вы найдете очень поддерживающее и полезное сообщество, поэтому, пожалуйста, заходите и расскажите нам, чем вы занимались. Или попробуйте ответить на вопросы тех, кто находится на более ранних этапах своего пути.
