Каждый из этих слоев получает входные данные от предыдущих слоев и постепенно уточняет их. Слои обучаются с помощью алгоритмов, которые минимизируют ошибки и повышают точность. Глубокое обучение обладает мощью, гибкостью и простотой. Именно поэтому мы считаем, что его следует применять в самых разных областях. ## Нейронные сети: Краткая история.


В 1943 году Уоррен Мак-Кулох, нейрофизиолог, и Уолтер Питтс, логик, объединили усилия для разработки математической модели искусственного нейрона. В своей [статье](https://link.springer.com/article/10.1007/BF02478259) "Логический исчисление, отражающее идеи, присущие нервной деятельности", они заявили следующее:

> : Благодаря "все или ничего" характеру нервной активности, нейронные процессы и отношения между ними могут быть описаны с помощью логики высказываний. Было установлено, что поведение любой нейронной сети можно описать в этих терминах.

Маккалох и Питтс осознали, что упрощенную модель реального нейрона можно представить с помощью простых операций сложения и пороговой обработки, как показано на схеме <<нейрон>>. Питтс был самоучкой, и уже в 12 лет ему было предложено учиться в Кембриджском университете у великого Бертрана Рассела. Он не принял это предложение, и, впрочем, на протяжении всей своей жизни не соглашался ни на какие предложения о получении ученых степеней или занятия руководящих должностей. Большая часть его известных работ была выполнена в период, когда он был бездомным. Несмотря на отсутствие официальной должности и растущую социальную изоляцию, его работа с Маккалохом оказала значительное влияние и была подхвачена психологом по имени Фрэнк Розенблатт.

```
<img alt="Естественные и искусственные нейроны" width="500" caption="Естественные и искусственные нейроны" src="images/chapter7_neuron.png" id="neuron"/>
```

Розенблатт продолжил совершенствовать искусственный нейрон, наделяя его способностью к обучению. Что еще более важно, он работал над созданием первого устройства, которое действительно использовало эти принципы – Mark I Perceptron. В работе "The Design of an Intelligent Automaton" Розенблатт писал об этой работе: "Мы стоим на пороге создания такой машины – машины, способной воспринимать, распознавать и идентифицировать окружающую среду без какого-либо человеческого обучения или контроля". Перцептрон был создан и успешно распознавал простые формы.

Профессор Массачусетского технологического института, Марвин Мински (который учился на год младше Розенблатта в одной и той же школе!), вместе с Сеймуром Пейпертом, написали книгу под названием _Perceptrons_ (издательство MIT Press), посвященную изобретению Розенблатта. Они показали, что один слой этих устройств не способен изучать некоторые простые, но важные математические функции (например, XOR). В той же книге они также показали, что использование нескольких слоев этих устройств позволит преодолеть эти ограничения. К сожалению, только первое из этих наблюдений получило широкое признание. В результате, мировое академическое сообщество практически полностью отказалось от нейронных сетей на следующие два десятилетия.

Возможно, одним из самых важных трудов в области нейронных сетей за последние 50 лет является многотомное издание *Parallel Distributed Processing* (PDP), написанное Дэвидом Румельхартом, Джеймсом Макклелланом и группой исследователей PDP, опубликованное в 1986 году издательством MIT Press. В первой главе авторы выражают надежду, схожую с той, что была высказана Розенблаттом:

> : Люди умнее, чем сегодняшние компьютеры, потому что мозг использует базовую вычислительную архитектуру, которая лучше подходит для решения ключевых задач обработки информации, в которых люди так хороши. ...Мы представим вычислительную модель для моделирования когнитивных процессов, которая, кажется, ближе, чем другие модели, к тому, как вычисления могут осуществляться в мозге.

Основная идея, которую PDP пытается донести, заключается в том, что традиционные компьютерные программы работают совершенно иначе, чем мозг, и именно поэтому компьютерные программы, по состоянию на тот момент, были настолько плохи в задачах, которые мозг решает легко (например, распознавание объектов на изображениях). Авторы утверждали, что подход PDP "ближе, чем другие модели", к тому, как работает мозг, и, следовательно, он может лучше справляться с такими задачами.

Фактически, подход, изложенный в PDP, очень похож на подход, используемый в современных нейронных сетях. В книге параллельная распределенная обработка определяется как требующая:

1. Набора *вычислительных единиц*
2. *Состояния активации*
3. *Функции вывода* для каждой единицы
4. *Паттерна связности* между единицами
5. *Правила распространения* для распространения паттернов активности по сети связей
6. *Правила активации* для объединения входных сигналов, поступающих на единицу, с текущим состоянием этой единицы для получения выходного сигнала
7. *Правила обучения*, посредством которых паттерны связности изменяются под воздействием опыта
8. *Окружающей среды*, в которой система должна функционировать.

В этой книге мы увидим, что современные нейронные сети решают все эти задачи.

В 1980-х годах большинство моделей строились с использованием второго слоя нейронов, что позволяло избежать проблемы, выявленной Минским и Папертом (это был их "паттерн связности между единицами", используя вышеуказанную структуру). И действительно, нейронные сети широко использовались в 1980-х и 1990-х годах для реальных, практических проектов. Однако, опять же, неправильное понимание теоретических аспектов тормозило развитие этой области. В теории, добавление всего одного дополнительного слоя нейронов было достаточно, чтобы любой математической функции можно было аппроксимировать с помощью этих нейронных сетей, но на практике такие сети часто были слишком большими и слишком медленными, чтобы быть полезными.

Хотя исследователи 30 лет назад показали, что для достижения практических результатов необходимо использовать еще больше слоев нейронов, только в последнее десятилетие этот принцип получил более широкое признание и применение. Нейронные сети, наконец, начинают реализовывать свой потенциал, благодаря использованию большего количества слоев, а также благодаря возможностям, появившимся благодаря улучшениям в аппаратном обеспечении компьютеров, увеличению объема доступных данных и алгоритмическим улучшениям, которые позволяют обучать нейронные сети быстрее и проще. Теперь у нас есть то, что Розенблат обещал: "машина, способная воспринимать, распознавать и идентифицировать свое окружение без какого-либо человеческого обучения или контроля".

Именно этому вы научитесь в этой книге. Но прежде, чем мы начнем, давайте немного познакомимся друг с другом…

## Как изучать глубокое обучение


Профессор Гарвардского университета Дэвид Перкинс, автор книги "Making Learning Whole" (издательство Jossey-Bass), много говорит об обучении. Основная идея заключается в том, чтобы обучать *целостному пониманию*. Это означает, что, например, если вы преподаете бейсбол, вы сначала водите людей на бейсбольные матчи или предлагаете им поиграть в него. Вы не учите их, как изготовить бейсбольный мяч с нуля, изучать физику параболы или коэффициент трения мяча о биту.

Пол Локхарт, доктор математики Колумбийского университета, бывший профессор Браунского университета и учитель математики в начальных и средних классах, в своей влиятельной [статье](https://www.maa.org/external_archive/devlin/LockhartsLament.pdf) "A Mathematician's Lament" описывает кошмарный мир, где музыку и искусство преподают так же, как математику. Детям не разрешается слушать или играть музыку, пока они не проведут более десяти лет, осваивая нотную грамоту и теорию, тратя время на переложение нот в другую тональность. 3.  Предварительно обученная модель будет *дообучена* с использованием последних достижений в области трансферного обучения, чтобы создать модель, специально адаптированную для распознавания собак и кошек.

Первые два этапа необходимо выполнить только один раз на вашем сервере с графическим процессором. Если вы запустите ячейку повторно, она будет использовать набор данных и модель, которые уже были загружены, а не загружать их снова. Давайте посмотрим, что содержится в ячейке, и на результаты (<<first_training>>):


````
#id first_training
#caption Results from the first training
# CLICK ME
from fastai.vision.all import *
path = untar_data(URLs.PETS)/'images'

def is_cat(x): return x[0].isupper()
dls = ImageDataLoaders.from_name_func(
    path, get_image_files(path), valid_pct=0.2, seed=42,
    label_func=is_cat, item_tfms=Resize(224))

learn = vision_learner(dls, resnet34, metrics=error_rate)
learn.fine_tune(1)
````

### Что такое машинное обучение?


Ваш классификатор – это модель глубокого обучения. Как уже упоминалось, модели глубокого обучения используют нейронные сети, которые впервые появились в 1950-х годах и стали мощными лишь в последнее время благодаря недавним достижениям.

Еще один важный аспект заключается в том, что глубокое обучение – это всего лишь современная область в более широкой дисциплине, известной как *машинное обучение*. Чтобы понять суть того, что вы сделали, когда обучали свою модель классификации, вам не обязательно разбираться в глубоком обучении. Достаточно понять, как ваша модель и процесс ее обучения являются примерами концепций, применимых к машинному обучению в целом.

В этом разделе мы опишем, что такое машинное обучение. Мы рассмотрим ключевые концепции и покажем, как их можно проследить до оригинальной работы, в которой они были впервые представлены.

*Машинное обучение*, как и обычное программирование, – это способ заставить компьютеры выполнять определенную задачу. Но как бы мы использовали обычное программирование для выполнения того, что мы делали в предыдущем разделе: распознавания собак и кошек на фотографиях? Нам бы пришлось записать для компьютера точные шаги, необходимые для выполнения задачи.

Обычно нам довольно легко записать шаги для выполнения задачи, когда мы пишем программу. Мы просто думаем о шагах, которые мы бы предприняли, если бы нам пришлось выполнять задачу вручную, а затем переводим их в код. Например, мы можем написать функцию, которая сортирует список. В общем, мы бы написали функцию, которая выглядит примерно так: <<basic_program>> (где *входные данные* могут быть несортированным списком, а *результат* – отсортированным списком).

````
#hide_input
#caption A traditional program
#id basic_program
#alt Pipeline inputs, program, results
gv('''program[shape=box3d width=1 height=0.7]
inputs->program->results''')
````

Однако распознавание объектов на фотографии – задача не из простых. Какие именно шаги мы предпринимаем, когда узнаем объект на изображении? Мы на самом деле не знаем, поскольку весь этот процесс происходит в нашем мозге, и мы не осознаем его!

Еще в начале эры компьютеров, в 1949 году, исследователь IBM по имени Артур Сэмюэл начал работать над новым способом, позволяющим компьютерам выполнять задачи, который он назвал "машинным обучением". В своем классическом эссе 1962 года "Искусственный интеллект: рубеж автоматизации" он написал:


Вот перевод:

> Программирование компьютера для выполнения таких вычислений – это, мягко говоря, сложная задача, и причина этого заключается не столько в какой-либо присущей сложности самого компьютера, сколько в необходимости детально прописывать каждый, даже самый незначительный, этап процесса. Компьютеры, как любой программист вам скажет, – это гигантские "тупицы", а не гигантские "мозги".

Его основная идея заключалась в следующем: вместо того, чтобы сообщать компьютеру точные шаги, необходимые для решения задачи, нужно показывать ему примеры задач и позволить ему самостоятельно найти способ их решения. Оказалось, что это очень эффективно: к 1961 году его программа для игры в шашки настолько усовершенствовалась, что победила чемпиона штата Коннектикут! Вот как он описал свою идею (из той же статьи, что и выше):


> Предположим, мы разработаем автоматизированный способ оценки эффективности текущей системы присвоения весов с точки зрения фактических результатов, и предоставим механизм для изменения этих весов таким образом, чтобы максимизировать производительность. Нам не нужно вдаваться в детали такого процесса, чтобы понять, что его можно сделать полностью автоматизированным, и что машина, запрограммированная таким образом, "научится" на своем опыте.

В этом коротком утверждении содержится ряд важных концепций:

- Идея "назначения весов".
- Тот факт, что каждое назначение весов имеет определенную "фактическую производительность".
- Требование наличия "автоматизированного способа" проверки этой производительности.
- Необходимость наличия "механизма" (то есть, другого автоматизированного процесса) для повышения производительности путем изменения назначений весов.

Рассмотрим эти концепции по порядку, чтобы понять, как они взаимодействуют на практике. Прежде всего, нам необходимо понять, что Сэмюэль подразумевает под *назначением весов*.

Веса – это просто переменные, а назначение весов – это конкретный выбор значений для этих переменных. Входные данные программы – это значения, которые она обрабатывает для получения результатов, например, принимая значения пикселей изображения и возвращая классификацию "собака". Назначения весов – это другие значения, которые определяют, как программа будет работать.

Поскольку они влияют на работу программы, их можно рассматривать как еще один вид входных данных. Поэтому мы изменим нашу базовую схему, представленную в разделе <<basic_program>>, и заменим ее схемой, иллюстрирующей <<weight_assignment>>, чтобы учесть это.

````
#hide_input
#caption A program using weight assignment
#id weight_assignment
gv('''model[shape=box3d width=1 height=0.7]
inputs->model->results; weights->model''')
````

Мы изменили название нашего блока с *program* (программа) на *model* (модель). Это сделано для соответствия современной терминологии и для того, чтобы отразить, что *модель* – это особый вид программы: это программа, которая может выполнять *многие разные задачи*, в зависимости от *параметров*. Ее можно реализовать различными способами. Например, в программе для игры в шашки, разработанной Сэмуэлем, различные значения параметров приводили бы к различным стратегиям игры.

(Кстати, то, что Сэмуэль называл "параметрами", в настоящее время чаще всего называют параметрами *модели*. Термин "параметры" зарезервирован для определенного типа параметров модели.)

Далее, Сэмуэль сказал, что нам необходимо *автоматизированный способ оценки эффективности любой текущей конфигурации параметров с точки зрения фактической производительности*. В случае его программы для игры в шашки, "фактическая производительность" модели определялась бы тем, насколько хорошо она играет. И можно было бы автоматически протестировать производительность двух моделей, заставив их играть друг против друга и наблюдая, какая из них обычно побеждает.

Наконец, он говорит, что нам нужен *механизм для изменения конфигурации параметров таким образом, чтобы максимизировать производительность*. Например, мы могли бы анализировать разницу в параметрах между выигрышной моделью и проигрышной моделью и немного корректировать параметры в сторону выигрышной модели.

Теперь мы можем понять, почему он сказал, что такую процедуру *можно сделать полностью автоматической, и... машина, запрограммированная таким образом, "обучалась бы" на своем опыте*. Обучение стало бы полностью автоматическим, когда корректировка параметров также была бы автоматической: когда вместо того, чтобы мы улучшали модель, вручную изменяя ее параметры, мы бы полагались на автоматизированный механизм, который производил бы корректировки на основе производительности.

<<training_loop>> показывает полную картину идеи Сэмуэля о том, как обучать модель машинного обучения.


````
#hide_input
#caption Training a machine learning model
#id training_loop
#alt The basic training loop
gv('''ordering=in
model[shape=box3d width=1 height=0.7]
inputs->model->results; weights->model; results->performance
performance->weights[constraint=false label=update]''')
````

### Что такое нейронная сеть?


Представьте себе, как могла бы выглядеть модель для программы, играющей в шашки. В ней могли бы быть заложены различные стратегии игры, механизм поиска оптимальных ходов, а веса могли бы определять, какие стратегии выбираются, на какие области доски делается акцент во время поиска и так далее. Однако, совершенно не очевидно, как могла бы выглядеть модель для программы распознавания изображений, для понимания текста или для решения многих других интересных задач, которые мы можем себе представить.

Нам нужна функция, настолько гибкая, что ее можно было бы использовать для решения любой поставленной задачи, просто изменяя ее параметры. Удивительно, но такая функция действительно существует! Это нейронная сеть, о которой мы уже говорили. То есть, если рассматривать нейронную сеть как математическую функцию, то это оказывается функция, обладающая чрезвычайной гибкостью, зависящей от ее параметров. Математическое доказательство, известное как *теорема об универсальной аппроксимации*, показывает, что эта функция теоретически может решить любую задачу с любой степенью точности. Тот факт, что нейронные сети обладают такой гибкостью, означает, что на практике они часто являются подходящей моделью, и вы можете сосредоточить свои усилия на процессе обучения этих сетей, то есть на подборе оптимальных значений параметров.

Но что касается этого процесса? Можно предположить, что для каждой задачи потребуется разрабатывать новый "механизм" для автоматической корректировки параметров. Это было бы очень трудоемко. Нам также нужен абсолютно общий способ обновления параметров нейронной сети, чтобы она могла улучшать свои результаты при выполнении любой задачи. К счастью, и это тоже существует!

Это называется *стохастическим градиентным спуском* (SGD). Мы подробно рассмотрим, как работают нейронные сети и SGD, в главе <<chapter_mnist_basics>>, а также объясним теорему об универсальной аппроксимации. Однако, пока мы воспользуемся словами самого Сэмуэля: *Нам не нужно вдаваться в детали такого процесса, чтобы понять, что его можно сделать полностью автоматическим, и что машина, запрограммированная таким образом, "учится" на своем опыте.*

Нейронные сети являются особенными, потому что они обладают высокой гибкостью, что означает, что они могут решать необычайно широкий спектр задач, просто находя правильные веса. ### Несколько терминов из области глубокого обучения


Сэмюэл работал в 1960-х годах, и с тех пор терминология изменилась. Вот современная терминология глубокого обучения для всех элементов, которые мы обсуждали:

- Функциональная форма *модели* называется ее *архитектурой* (но будьте осторожны: иногда люди используют слово *модель* как синоним *архитектуры*, что может привести к путанице).
- *Веса* называются *параметрами*.
- *Прогнозы* рассчитываются на основе *независимой переменной*, которая является *данными*, не включающими *метки*.
- *Результаты* работы модели называются *прогнозами*.
- Мера *производительности* называется *функцией потерь*.
- Функция потерь зависит не только от прогнозов, но и от правильных *меток* (также известных как *целевые значения* или *зависимая переменная*); например, "собака" или "кошка".

### Ограничения, присущие машинному обучению

Из этого описания мы можем выделить несколько ключевых моментов, касающихся обучения модели глубокого обучения:

- Модель не может быть создана без данных.
- Модель может обучаться только на основе закономерностей, присутствующих во входных данных, используемых для ее обучения.
- Этот подход к обучению позволяет делать только *прогнозы*, а не рекомендовать *действия*.
- Недостаточно просто иметь примеры входных данных; нам также нужны *метки* для этих данных (например, изображений собак и кошек недостаточно для обучения модели; для каждого изображения требуется метка, указывающая, является ли оно изображением собаки или кошки).

В целом, мы видим, что большинство организаций, утверждающих, что у них недостаточно данных, на самом деле имеют недостаточно *размеченных* данных. Если какая-либо организация заинтересована в практическом применении модели, то, вероятно, у нее есть какие-то входные данные, на которых она планирует протестировать свою модель. И, вероятно, они уже какое-то время используют другие методы для обработки этих данных (например, вручную или с помощью некоторой эвристической программы), поэтому у них уже есть данные, полученные в результате этих процессов! Например, в радиологической клинике, скорее всего, есть архив медицинских снимков (поскольку необходимо отслеживать прогресс лечения пациентов), но эти снимки могут не иметь структурированных меток, содержащих список диагнозов или проведенных процедур (поскольку радиологи обычно составляют отчеты на естественном языке, а не структурированные данные). Мы будем подробно рассматривать методы разметки данных в этой книге, поскольку это очень важный вопрос на практике.

Поскольку такие модели машинного обучения могут делать только *прогнозы* (то есть пытаться воспроизвести метки), это может привести к существенному разрыву между целями организации и возможностями модели. Например, в этой книге вы узнаете, как создать *систему рекомендаций*, которая может предсказывать, какие продукты пользователь может приобрести. Это часто используется в электронной коммерции, например, для персонализации продуктов, отображаемых на главной странице, путем показа наиболее популярных товаров. Однако такая модель обычно создается путем анализа истории покупок пользователя (*входные данные*) и того, что он приобретал или просматривал (*метки*), что означает, что модель, скорее всего, будет предлагать пользователю продукты, которые у него уже есть или о которых он уже знает, а не новые продукты, которые ему, вероятно, было бы интересно узнать. Это очень отличается от того, что, например, может сделать эксперт в вашем местном книжном магазине, который задает вопросы, чтобы понять ваши предпочтения, а затем рассказывает вам об авторах или сериях, о которых вы раньше не слышали.

Еще одно важное понимание приходит при рассмотрении того, как модель взаимодействует со своей средой. Это может создавать *петли обратной связи*, как описано здесь:

- Модель прогнозирования преступности создается на основе данных о местах, где в прошлом были произведены аресты. На практике это не является прогнозированием преступлений, а скорее прогнозированием арестов, и, следовательно, частично просто отражает предвзятость, существующую в текущих процессах работы полиции.
- Сотрудники правоохранительных органов могут использовать эту модель, чтобы определить, где сосредоточить свою деятельность, что приводит к увеличению числа арестов в этих районах.
- Данные о дополнительных арестах затем используются для повторной настройки будущих версий модели.

Это *положительная петля обратной связи*, при которой чем больше модель используется, тем более предвзятыми становятся данные, что делает модель еще более предвзятой, и так далее.

Петли обратной связи также могут создавать проблемы в коммерческой сфере. Например, система рекомендаций видео может быть предвзятой в отношении контента, который потребляют самые активные пользователи (например, сторонники теорий заговора и экстремисты, как правило, смотрят больше онлайн-видео, чем в среднем), что приводит к тому, что эти пользователи увеличивают потребление видео, что, в свою очередь, приводит к тому, что им предлагается все больше видео такого рода. Мы рассмотрим эту тему более подробно в главе <<chapter_ethics>>.

Это два основных типа моделей, которые мы будем изучать в этой книге. Модель классификации – это модель, которая пытается предсказать класс или категорию. То есть, она предсказывает один из множества дискретных вариантов, например, "собака" или "кошка". Модель регрессии – это модель, которая пытается предсказать одно или несколько числовых значений, например, температуру или географические координаты. Иногда люди используют слово _регрессия_ для обозначения определенного типа модели, называемой _линейной регрессионной моделью_; это неверная практика, и мы не будем использовать эту терминологию в этой книге!

Набор данных Pet содержит 7390 изображений собак и кошек, представляющих 37 различных пород. Каждое изображение имеет метку, указанную в его имени файла: например, файл *great\_pyrenees\_173.jpg* является 173-м примером изображения собаки породы Great Pyrenees в этом наборе данных. Имена файлов начинаются с заглавной буквы, если на изображении кошка, и со строчной буквы в противном случае. Нам нужно сообщить библиотеке fastai, как извлекать метки из имен файлов. Мы делаем это, вызывая функцию `from_name_func` (что означает, что метки могут быть извлечены с помощью функции, применяемой к имени файла), и передавая параметр `is_cat`, который возвращает `x[0].isupper()`. Это выражение возвращает `True`, если первая буква в имени файла является заглавной (то есть, это кошка).

Наиболее важным параметром, который стоит упомянуть, является `valid_pct=0.2`. Этот параметр указывает библиотеке fastai, что необходимо выделить 20% данных и *не использовать их для обучения модели вообще*. Эти 20% данных называются *набором для валидации*; оставшиеся 80% называются *набором для обучения*. Набор для валидации используется для оценки точности модели. По умолчанию, эти 20%, которые выделяются, выбираются случайным образом. Параметр `seed=42` задает *случайное зерно* для получения одинаковой последовательности случайных чисел при каждом запуске этого кода. Это означает, что мы всегда будем получать один и тот же набор для валидации при каждом запуске. Таким образом, если мы изменим нашу модель и переобучим ее, мы будем уверены, что любые различия связаны с изменениями в самой модели, а не с использованием другого случайного набора для валидации.

Библиотека fastai *всегда* показывает вам точность вашей модели, используя *только* набор для валидации, *никогда* набор для обучения. Это абсолютно критически важно, потому что, если обучать достаточно большую модель достаточно долго, она в конечном итоге запомнит метку каждого элемента в вашем наборе данных! В результате, мы получим не полезную модель, потому что нас интересует, насколько хорошо наша модель работает на *изображениях, которые она никогда не видела раньше*. Это всегда является нашей целью при создании модели: чтобы она была полезна для данных, которые модель увидит только в будущем, после того, как она будет обучена.

Даже если ваша модель еще не полностью запомнила все данные, в начале обучения она могла запомнить определенные ее части. В результате, чем дольше вы обучаете модель, тем лучше будет ее точность на наборе для обучения; точность на наборе для валидации также будет улучшаться некоторое время, но в конечном итоге она начнет ухудшаться, когда модель начнет запоминать набор для обучения, а не находить общие закономерности в данных. Когда это происходит, мы говорим, что модель *переобучена*.

<<img_overfit>> показывает, что происходит при переобучении, используя упрощенный пример, в котором у нас есть только один параметр и некоторые случайно сгенерированные данные на основе функции `x**2`. Как вы можете видеть, хотя прогнозы в переобученной модели точны для данных, близких к наблюдаемым точкам, они сильно неверны, когда находятся вне этого диапазона.

**Переобучение – это, пожалуй, самая важная и сложная проблема**, с которой сталкиваются все специалисты в области машинного обучения и все алгоритмы. Как вы увидите, очень легко создать модель, которая отлично справляется с предсказаниями на тех данных, на которых она была обучена, но гораздо сложнее делать точные предсказания на данных, которые модель никогда раньше не видела. И, конечно, именно эти данные будут иметь значение на практике. Например, если вы создаете классификатор рукописных цифр (о чем мы скоро поговорим!) и используете его для распознавания чисел, написанных на чеках, то вы никогда не увидите тех чисел, на которых модель была обучена, поскольку на чеках будут использоваться немного отличающиеся варианты написания. В этой книге вы узнаете множество методов, позволяющих избежать переобучения. Однако, вы должны использовать эти методы только после того, как убедитесь, что переобучение действительно происходит (то есть, вы наблюдаете, что точность модели на проверочных данных ухудшается в процессе обучения). Часто бывает так, что специалисты используют методы предотвращения переобучения, даже когда у них достаточно данных, и это приводит к созданию модели, которая может быть менее точной, чем могла бы быть.

Внимание: Набор данных для валидации: При обучении модели необходимо *всегда* иметь как обучающий набор данных, так и набор данных для валидации, и оценивать точность вашей модели только на наборе данных для валидации. Если вы обучаете модель слишком долго и с недостаточным количеством данных, вы увидите, что точность модели начинает ухудшаться; это называется *переобучением*. По умолчанию, в fastai для параметра `valid_pct` установлено значение `0.2`, поэтому даже если вы забудете, fastai автоматически создаст набор данных для валидации.

Пятая строка кода, используемая для обучения нашей системы распознавания изображений, указывает библиотеке fastai создать *сверточную нейронную сеть* (CNN) и определяет *архитектуру*, которую следует использовать (то есть, какой тип модели создать), данные, на которых мы хотим ее обучить, и *метрику*, которую необходимо использовать:

```python
learn = vision_learner(dls, resnet34, metrics=error_rate)
```

Почему CNN? Это современный, передовой подход к созданию моделей компьютерного зрения. В этой книге мы подробно рассмотрим, как работают CNN. Их структура вдохновлена тем, как работает человеческая зрительная система.

В fastai существует множество различных архитектур, которые мы представим в этой книге (а также обсудим, как создать свои собственные). Однако, в большинстве случаев, выбор архитектуры не является самым важным аспектом процесса глубокого обучения. Это тема, которую любят обсуждать ученые, но на практике, скорее всего, вам не придется тратить на это много времени. Существуют стандартные архитектуры, которые хорошо работают в большинстве случаев, и в данном случае мы используем одну из них, называемую _ResNet_, о которой мы будем много говорить в течение всей книги; она достаточно быстра и точна для многих наборов данных и задач. Число `34` в `resnet34` указывает на количество слоев в этой версии архитектуры (другие варианты: `18`, `50`, `101` и `152`). Модели, использующие архитектуры с большим количеством слоев, требуют больше времени для обучения и более подвержены переобучению (то есть, их нельзя обучать слишком долго, прежде чем точность на проверочном наборе начнет ухудшаться). С другой стороны, при использовании большего объема данных, они могут быть значительно более точными.

Что такое метрика? *Метрика* — это функция, которая измеряет качество предсказаний модели, используя проверочный набор данных, и которая будет отображаться в конце каждой *эпохи*. В данном случае мы используем `error_rate`, которая является функцией, предоставляемой fastai, и которая, как следует из названия, показывает процент изображений в проверочном наборе данных, которые классифицируются неправильно. Другая распространенная метрика для классификации — `accuracy` (которая просто равна `1.0 - error_rate`). Fastai предоставляет множество других метрик, которые будут рассмотрены на протяжении всей книги.

Концепция метрики может напомнить вам о *функции потерь* (loss), но между ними есть важное различие. Единственная цель функции потерь — определить "показатель производительности", который система обучения может использовать для автоматической корректировки весов. Другими словами, хороший выбор функции потерь — это такой выбор, который легко использовать для стохастического градиентного спуска. Но метрика предназначена для использования человеком, поэтому хорошая метрика — это такая, которую вам легко понять, и которая максимально соответствует тому, что вы хотите, чтобы модель делала. Иногда вы можете решить, что функция потерь является подходящей метрикой, но это не всегда так.


`vision_learner` также имеет параметр `pretrained`, который по умолчанию установлен в `True` (то есть он используется в данном случае, даже если мы его явно не указали). Этот параметр задает веса в вашей модели значениями, которые уже были обучены экспертами для распознавания тысячи различных категорий на 1,3 миллионах фотографий (используя известный набор данных [*ImageNet*](http://www.image-net.org/)). Модель, веса которой уже были обучены на каком-либо другом наборе данных, называется *моделью, предварительно обученной*. Вам почти всегда следует использовать предварительно обученную модель, потому что это означает, что ваша модель, еще до того, как вы покажете ей какие-либо ваши данные, уже обладает значительными возможностями. И, как вы увидите, во многих моделях глубокого обучения эти возможности необходимы практически независимо от деталей вашего проекта. Например, части предварительно обученных моделей обеспечивают обнаружение краев, градиентов и цветов, что необходимо для многих задач.

При использовании предварительно обученной модели `vision_learner` удаляет последний слой, поскольку он всегда специально адаптирован к исходной задаче обучения (например, классификации набора данных ImageNet), и заменяет его одним или несколькими новыми слоями со случайными весами, подходящего размера для используемого вами набора данных. Эта последняя часть модели называется *головной частью* (head).

Использование предварительно обученных моделей – это *самый* важный метод, который позволяет нам обучать более точные модели, быстрее, с использованием меньшего объема данных, а также с меньшими затратами времени и денег. Возможно, вы подумаете, что это означает, что использование предварительно обученных моделей – это наиболее изученная область в академическом глубоком обучении... но вы будете очень, очень неправы! Важность предварительно обученных моделей, как правило, не признается и не обсуждается в большинстве курсов, книг или функций библиотек программного обеспечения, и редко рассматривается в научных публикациях. На момент написания этой статьи в начале 2020 года ситуация начинает меняться, но это, вероятно, займет некоторое время. Поэтому будьте осторожны: большинство людей, с которыми вы будете разговаривать, вероятно, сильно недооценят, что вы можете сделать в глубоком обучении с ограниченными ресурсами, потому что они, скорее всего, не будут глубоко понимать, как использовать предварительно обученные модели.

Использование предварительно обученной модели для задачи, отличной от той, для которой она была изначально обучена, называется *переносным обучением* (transfer learning). К сожалению, из-за того, что переносное обучение изучено недостаточно, в настоящее время доступно лишь небольшое количество предварительно обученных моделей для различных областей. Например, в настоящее время существует мало предварительно обученных моделей в медицине, что затрудняет использование переноса обучения в этой области. Кроме того, еще недостаточно хорошо изучено, как использовать переносное обучение для таких задач, как анализ временных рядов.

```
жаргон: Перенос обучения (transfer learning): Использование предварительно обученной модели для решения задачи, отличной от той, для которой она была изначально обучена.
```

Шестая строка нашего кода сообщает библиотеке fastai, как обучать модель:

```python
learn.fine_tune(1)
```

Как мы уже обсуждали, архитектура описывает лишь *шаблон* для математической функции; она не выполняет никаких действий, пока мы не предоставим значения для миллионов параметров, которые она содержит.

Это ключевой момент глубокого обучения: определение способа подстройки параметров модели, чтобы она решала вашу задачу. Для обучения модели необходимо предоставить хотя бы одну информацию: сколько раз нужно просматривать каждое изображение (это называется число *эпох*). Количество эпох, которое вы выберете, во многом будет зависеть от того, сколько времени у вас есть, и сколько времени требуется на практике для обучения вашей модели. Если вы выберете слишком маленькое число, вы всегда сможете обучить модель большее количество эпох позже.

Но почему этот метод называется `fine_tune`, а не `fit`? В библиотеке fastai действительно есть метод под названием `fit`, который, собственно, и предназначен для обучения модели (то есть, для многократного просмотра изображений в обучающем наборе, при каждом просмотре обновляя параметры, чтобы прогнозы становились все ближе и ближе к целевым меткам). Но в данном случае мы начинаем с предварительно обученной модели, и мы не хотим потерять все те возможности, которыми она уже обладает. Как вы узнаете из этой книги, существуют важные приемы для адаптации предварительно обученной модели к новому набору данных — это процесс, называемый *дообучением* (fine-tuning).

```
терминология: Тонкая настройка: Техника обучения с переносом, при которой параметры предварительно обученной модели корректируются путем дополнительного обучения в течение нескольких эпох, используя задачу, отличную от той, которая использовалась для предварительного обучения.
```

При использовании метода `fine_tune`, библиотека fastai автоматически применяет эти приемы. Существует несколько параметров, которые можно настроить (о чем мы поговорим позже), но в своей стандартной форме, представленной здесь, метод выполняет два этапа:

1.  Используется один эпох для обучения только тех частей модели, которые необходимы для правильной работы нового "головного" блока с вашим набором данных.
2.  Используется количество эпох, указанное при вызове метода, для обучения всей модели, при этом веса более поздних слоев (особенно "головного" блока) обновляются быстрее, чем веса более ранних слоев (которые, как мы увидим, обычно не требуют значительных изменений по сравнению с предварительно обученными весами).

*"Головной" блок модели – это часть, которая добавляется для адаптации модели к новому набору данных. *Эпоха* – это один полный проход по набору данных. После вызова метода `fit` выводятся результаты после каждой эпохи, показывающие номер эпохи, значения потерь для обучающего и валидационного наборов (мера производительности, используемая для обучения модели) и любые *метрики*, которые вы запросили (в данном случае, коэффициент ошибок).

Итак, благодаря всему этому коду, наша модель научилась распознавать кошек и собак, используя только размеченные примеры. Но как она это сделала?

### Что узнал наш модуль распознавания изображений.

На данном этапе у нас есть система распознавания изображений, которая работает очень хорошо, но мы совершенно не знаем, что она на самом деле делает! Хотя многие люди жалуются, что методы глубокого обучения приводят к непрозрачным "черным ящикам" (то есть, к системам, которые выдают прогнозы, но которые никто не может понять), это, на самом деле, далеко от истины. Существует огромный объем исследований, показывающих, как можно глубоко анализировать модели глубокого обучения и получать из них ценные сведения. Тем не менее, любые модели машинного обучения (включая глубокое обучение и традиционные статистические модели) могут быть сложными для полного понимания, особенно когда речь идет о том, как они будут вести себя при работе с данными, которые сильно отличаются от данных, использованных для их обучения. Мы будем обсуждать этот вопрос на протяжении всей этой книги.

В 2013 году аспирант Мэтт Зейлер и его научный руководитель Роб Фергус опубликовали статью ["Визуализация и понимание сверточных нейронных сетей"](https://arxiv.org/pdf/1311.2901.pdf), в которой было показано, как визуализировать веса нейронной сети, которые усваиваются на каждом слое модели. Они тщательно проанализировали модель, победившую в конкурсе ImageNet в 2012 году, и использовали этот анализ для значительного улучшения модели, что позволило им победить в конкурсе 2013 года! <<img_layer1>> - это изображение, которое они опубликовали, показывающее веса первого слоя.

Теперь, когда у нас есть этот набор терминов, мы можем объединить все ключевые понятия, которые были представлены ранее. Пожалуйста, уделите время для повторного ознакомления с определениями и прочитайте следующее резюме. Если вы сможете понять объяснения, то будете хорошо подготовлены к дальнейшим обсуждениям.

*Машинное обучение* — это область, в которой мы определяем программу не путем ее полной разработки, а путем обучения на основе данных. *Глубокое обучение* — это специализированная область машинного обучения, которая использует *нейронные сети* с несколькими *слоями*. *Классификация изображений* — это типичный пример (также известный как *распознавание изображений*). Мы начинаем с *размеченных данных*; то есть, с набора изображений, к каждому из которых мы присвоили *метку*, указывающую на то, что на нем изображено. Наша цель — создать программу, называемую *моделью*, которая, получив новое изображение, сделает точный *прогноз* относительно того, что на этом изображении представлено.

Каждая модель начинается с выбора *архитектуры*, которая представляет собой общую структуру, определяющую, как эта модель работает внутри. Процесс *обучения* (или *подгонки*) модели заключается в поиске набора *значений параметров* (или *весов*), которые адаптируют эту общую архитектуру к модели, которая хорошо работает для нашего конкретного типа данных. Чтобы определить, насколько хорошо модель справляется с отдельным прогнозом, нам необходимо определить *функцию потерь*, которая определяет, как мы оцениваем прогноз как хороший или плохой.

Чтобы ускорить процесс обучения, мы можем начать с *предварительно обученной модели* — модели, которая уже была обучена на данных, принадлежащих другим. Затем мы можем адаптировать ее к нашим данным, обучая ее еще немного на наших данных, что называется *тонкой настройкой*.

При обучении модели одним из ключевых моментов является обеспечение того, чтобы наша модель *обобщала* — то есть, чтобы она извлекала общие закономерности из наших данных, которые также применимы к новым элементам, с которыми она столкнется, чтобы она могла делать точные прогнозы относительно этих элементов. Риск заключается в том, что, если мы плохо обучим нашу модель, вместо того чтобы извлекать общие закономерности, она фактически запоминает то, что уже видела, и тогда она будет делать неточные прогнозы относительно новых изображений. Такой сбой называется *переобучением*. Чтобы избежать этого, мы всегда делим наши данные на две части: *обучающий набор* и *набор для проверки*. Мы обучаем модель, показывая ей только обучающий набор, а затем оцениваем, насколько хорошо она работает, анализируя ее производительность на элементах из набора для проверки. Таким образом, мы проверяем, являются ли уроки, которые модель извлекает из обучающего набора, уроками, которые применимы к набору для проверки. Чтобы человек мог оценить, насколько хорошо модель работает на наборе для проверки в целом, мы определяем *метрику*. Во время процесса обучения, когда модель видит все элементы обучающего набора, мы называем это *эпохой*.

Все эти концепции применимы к машинному обучению в целом. То есть, они применимы ко всем видам схем для определения модели путем ее обучения на данных. То, что делает глубокое обучение особенным, — это определенный класс архитектур: архитектуры, основанные на *нейронных сетях*. В частности, такие задачи, как классификация изображений, в значительной степени зависят от *сверточных нейронных сетей*, о которых мы поговорим позже.

## Глубокое обучение – это не только классификация изображений.


Эффективность глубокого обучения в задачах классификации изображений широко обсуждалась в последние годы, и даже демонстрируются результаты, превосходящие человеческие возможности, в сложных задачах, таких как распознавание злокачественных опухолей на компьютерной томографии. Однако, возможности глубокого обучения гораздо шире, и мы покажем это здесь.

Например, давайте поговорим о чем-то, что критически важно для автономных транспортных средств: локализация объектов на изображении. Если беспилотный автомобиль не знает, где находится пешеход, он не знает, как избежать столкновения! Создание модели, которая может распознавать содержимое каждого отдельного пикселя на изображении, называется *сегментацией*. Вот как мы можем обучить модель сегментации с помощью библиотеки fastai, используя подмножество набора данных [*Camvid*](http://www0.cs.ucl.ac.uk/staff/G.Brostow/papers/Brostow_2009-PRL.pdf) из статьи "Semantic Object Classes in Video: A High-Definition Ground Truth Database", написанной Габриэлем Дж. Бростоу, Жюльеном Фокье и Роберто Чиполлой:


````
path = untar_data(URLs.CAMVID_TINY)
dls = SegmentationDataLoaders.from_label_func(
    path, bs=8, fnames = get_image_files(path/"images"),
    label_func = lambda o: path/'labels'/f'{o.stem}_P{o.suffix}',
    codes = np.loadtxt(path/'codes.txt', dtype=str)
)

learn = unet_learner(dls, resnet34)
learn.fine_tune(8)
````

Мы даже не будем разбирать этот код построчно, потому что он почти идентичен нашему предыдущему примеру! (Хотя мы подробно рассмотрим модели сегментации в разделе «<<chapter_arch_details>>», а также все остальные модели, которые мы кратко представляем в этой главе, и многие другие.)

Мы можем визуализировать, насколько хорошо модель выполняет свою задачу, попросив ее раскрасить каждый пиксель изображения разными цветами. Как видите, модель почти идеально классифицирует каждый пиксель в каждом объекте. ```python
````
from fastai.text.all import *

dls = TextDataLoaders.from_folder(untar_data(URLs.IMDB), valid='test')
````

```
Терминология: Табличные данные: Данные, представленные в виде таблицы, например, из электронной таблицы, базы данных или файла CSV. Табличная модель – это модель, которая пытается предсказать значение одного столбца таблицы, основываясь на информации, содержащейся в других столбцах этой же таблицы.
```

Оказывается, это выглядит очень похоже. Вот код, необходимый для обучения модели, которая будет предсказывать, относится ли человек к категории людей с высоким уровнем дохода, основываясь на его социально-экономическом положении:

Как вы видите, нам пришлось сообщить библиотеке fastai, какие столбцы являются *категориальными* (то есть содержат значения, которые являются одним из предопределенного набора вариантов, например, `occupation`), а какие являются *непрерывными* (то есть содержат число, представляющее величину, например, `age`).

Для этой задачи нет готовой предварительно обученной модели (в целом, предварительно обученные модели не так широко доступны для задач работы с табличными данными, хотя некоторые организации создали их для внутреннего использования), поэтому мы не используем `fine_tune` в данном случае. Вместо этого мы используем `fit_one_cycle`, наиболее распространенный метод обучения моделей fastai *с нуля* (то есть без использования трансферного обучения):

````
learn.fit_one_cycle(3)
````

Эта модель использует набор данных [*Adult*](http://robotics.stanford.edu/~ronnyk/nbtree.pdf), представленный в статье "Повышение точности классификаторов Naive-Bayes: гибридный подход с использованием деревьев решений", автором которой является Роб Кохави. Точность модели составляет более 80%, а время ее обучения составило около 30 секунд.

## Наборы данных для проверки и тестовые наборы


Как мы уже обсуждали, цель модели – делать прогнозы на основе данных. Однако процесс обучения модели по своей сути является довольно простым. Если бы мы обучили модель на всех наших данных, а затем оценили ее производительность, используя те же самые данные, мы бы не смогли понять, насколько хорошо модель может работать с данными, которые она не видела. Без этой очень важной информации, которая должна направлять нас в процессе обучения модели, существует большая вероятность того, что модель хорошо будет справляться с прогнозами на основе тех данных, на которых она была обучена, но будет плохо работать с новыми данными.

Чтобы избежать этого, нашим первым шагом было разделить наш набор данных на два набора: *обучающий набор* (на котором модель обучается) и *набор для валидации*, также известный как *набор для разработки* (который используется только для оценки). Это позволяет нам проверить, усваивает ли модель уроки из обучающих данных, которые применимы к новым данным, то есть к данным из набора валидации.

Один из способов понять эту ситуацию заключается в том, что, в некотором смысле, мы не хотим, чтобы модель получала хорошие результаты путем "обмана". Если модель делает точный прогноз для определенного элемента данных, это должно быть связано с тем, что она изучила характеристики этого типа элементов, а не с тем, что модель была "настроена" *именно на этот конкретный элемент*.

Разделение данных на обучающий и валидационный наборы означает, что модель никогда не видит данные из набора валидации в процессе обучения, и, следовательно, она полностью свободна от их влияния и не "обманывает" ни в коем случае. Верно?

На самом деле, не обязательно. Ситуация более сложная. Это связано с тем, что в реальных сценариях мы редко создаем модель, просто обучая ее параметры один раз. Вместо этого, мы, скорее всего, будем исследовать множество версий модели, используя различные подходы к построению сети, скорости обучения, стратегиям увеличения объема данных и другим факторам, которые мы обсудим в следующих главах. Многие из этих выборов можно охарактеризовать как выбор *гиперпараметров*. Это слово отражает тот факт, что это параметры, относящиеся к другим параметрам, поскольку они являются параметрами более высокого уровня, которые определяют смысл параметров модели.

Проблема заключается в том, что, хотя в обычном процессе обучения модель оценивает свои прогнозы только на обучающих данных при определении значений параметров, это не относится к нам. Мы, как разработчики моделей, оцениваем модель, анализируя прогнозы на проверочных данных, когда решаем поэкспериментировать с новыми значениями гиперпараметров! Таким образом, последующие версии модели, косвенно, формируются под влиянием того, что мы видели проверочные данные. Как и автоматический процесс обучения может привести к переобучению на обучающих данных, мы можем столкнуться с переобучением на проверочных данных из-за человеческих ошибок и проб.

Решением этой проблемы является введение еще одного уровня данных, которые мы не используем: *тестовый набор*. Так же, как мы отключаем проверочные данные от процесса обучения, мы должны отделить и тестовый набор даже от себя. Его нельзя использовать для улучшения модели; он может использоваться только для оценки модели в самом конце нашей работы. Фактически, мы определяем иерархию разделения наших данных, основанную на том, насколько тщательно мы хотим скрыть их от процессов обучения и моделирования: обучающие данные полностью доступны, проверочные данные доступны в меньшей степени, а тестовые данные полностью скрыты. Эта иерархия отражает различные типы процессов моделирования и оценки: автоматический процесс обучения с обратным распространением ошибки, более ручной процесс подбора различных гиперпараметров между сессиями обучения и оценка наших окончательных результатов.

Тестовый и проверочный наборы должны содержать достаточно данных, чтобы обеспечить получение точной оценки вашей точности. Например, если вы создаете детектор кошек, вам, как правило, требуется не менее 30 изображений кошек в вашем проверочном наборе. Это означает, что если у вас есть набор данных, содержащий тысячи элементов, использование стандартного размера проверочного набора (20%) может быть избыточным. С другой стороны, если у вас очень много данных, использование части из них для проверки, вероятно, не имеет никаких недостатков.

Наличие двух уровней "резервных данных" — проверочного набора и тестового набора, где один уровень представляет данные, которые вы практически скрываете от себя, — может показаться несколько чрезмерным. Но причина, по которой это часто необходимо, заключается в том, что модели, как правило, стремятся к самому простому способу делать точные прогнозы (запоминанию), а мы, как несовершенные люди, склонны обманывать себя относительно того, насколько хорошо работают наши модели. Дисциплина, связанная с использованием тестового набора, помогает нам сохранять интеллектуальную честность. Это не означает, что нам *всегда* нужен отдельный тестовый набор — если у вас очень мало данных, вам может потребоваться только проверочный набор, — но, как правило, лучше использовать тестовый набор, если это возможно.

Этот же принцип может быть критически важным, если вы планируете нанять стороннюю компанию для выполнения работ по моделированию. Сторонняя компания может не полностью понимать ваши требования, или их мотивы могут даже подталкивать их к неправильному пониманию. Хорошо подготовленный тестовый набор может значительно снизить эти риски и позволит вам оценить, решает ли их работа вашу фактическую проблему.

Если говорить прямо, если вы являетесь руководителем в вашей организации (или консультируете руководителей), самое важное, что вам нужно понять: если вы действительно понимаете, что такое тестовые и проверочные наборы и почему они важны, вы избежите основной причины неудач, с которыми сталкиваются организации, решившие использовать искусственный интеллект. Например, если вы рассматриваете возможность привлечения стороннего поставщика или сервиса, убедитесь, что вы выделяете некоторые тестовые данные, которые этот поставщик *никогда не должен видеть*. Затем *вы* проверяете их модель на ваших тестовых данных, используя метрику, которую *вы* выбираете на основе того, что на самом деле важно для вас на практике, и *вы* определяете, какой уровень производительности является приемлемым. (Также полезно вам самим попробовать создать простую базовую модель, чтобы вы знали, чего может достичь даже очень простая модель. Часто оказывается, что ваша простая модель работает так же хорошо, как и модель, созданная внешним "экспертом"!)


### Проявляйте осмотрительность при определении тестовых наборов.


Чтобы правильно сформировать набор данных для валидации (и, возможно, тестовый набор), иногда необходимо сделать больше, чем просто случайным образом выбрать часть исходного набора данных. Помните: ключевым свойством наборов данных для валидации и тестирования является то, что они должны быть репрезентативными для новых данных, которые вы будете видеть в будущем. Это может показаться невыполнимой задачей! По определению, вы еще не видели эти данные. Но обычно вы все равно знаете кое-что.

Полезно рассмотреть несколько примеров. Многие из этих примеров взяты из соревнований по предиктивному моделированию на платформе [Kaggle](https://www.kaggle.com/), которая хорошо иллюстрирует проблемы и методы, с которыми вы можете столкнуться на практике.

В одном случае, например, вы работаете с данными временных рядов. Для данных временных рядов случайный выбор подмножества данных будет одновременно слишком простым (поскольку вы можете видеть данные как до, так и после дат, которые вы пытаетесь предсказать) и нерепрезентативным для большинства практических задач (где вы используете исторические данные для построения модели, предназначенной для использования в будущем). Если ваши данные содержат дату и вы строите модель для использования в будущем, вам следует выбрать непрерывный фрагмент с самыми последними датами в качестве набора данных для валидации (например, последние две недели или последний месяц доступных данных).

Предположим, вы хотите разделить данные временного ряда, обозначенные как <<timeseries1>>, на обучающий и валидационный наборы.


Вместо этого, используйте более ранние данные в качестве обучающего набора (а более поздние данные – для набора для проверки), как показано на рисунке <<timeseries3>>.

<img src="images/timeseries3.png" width="400" id="timeseries3" caption="Хороший набор данных для обучения" alt="Набор данных для обучения, использующий данные до определенного момента времени">

Например, на платформе Kaggle проводился конкурс по [прогнозированию объемов продаж сети продуктовых магазинов в Эквадоре](https://www.kaggle.com/c/favorita-grocery-sales-forecasting). Данные для обучения, предоставленные Kaggle, охватывали период с 1 января 2013 года по 15 августа 2017 года, а тестовые данные – период с 16 августа 2017 года по 31 августа 2017 года. Таким образом, организатор конкурса обеспечил, чтобы участники делали прогнозы для периода, который был *в будущем*, с точки зрения их моделей. Это похоже на то, как трейдеры количественных хедж-фондов проводят *обратное тестирование*, чтобы проверить, насколько их модели способны предсказывать будущие периоды, основываясь на исторических данных.

Второй распространенный случай возникает, когда вы можете легко предвидеть, что данные, на основе которых вы будете делать прогнозы в реальных условиях, могут *качественно отличаться* от данных, используемых для обучения вашей модели.

В конкурсе Kaggle [обнаружение отвлеченных водителей](https://www.kaggle.com/c/state-farm-distracted-driver-detection), независимыми переменными являются фотографии водителей, находящихся за рулем автомобиля, а зависимыми переменными – категории, такие как отправка текстовых сообщений, прием пищи или безопасное наблюдение за дорогой. На многих фотографиях изображены одни и те же водители в разных позах, как мы видим на изображении <<img_driver>>. Если бы вы были страховой компанией, создающей модель на основе этих данных, важно отметить, что вас больше всего интересует, насколько хорошо модель работает с водителями, которых она не видела раньше (поскольку, вероятно, у вас будут данные для обучения только для небольшой группы людей). Учитывая это, тестовые данные для конкурса состоят из изображений людей, которые не присутствуют в обучающем наборе.

<img src="images/driver.PNG" width="600" id="img_driver" caption="Два изображения из обучающего набора данных" alt="Два изображения из обучающего набора данных, на которых изображен один и тот же водитель">

Если вы поместите одно из изображений из каталога <<img_driver>> в обучающий набор и другое – в набор для проверки, вашей модели будет легко сделать прогноз для изображения из набора для проверки, поэтому она покажет лучшие результаты, чем если бы ее использовали для анализа изображений новых людей. Другая точка зрения заключается в том, что, если вы использовали только изображения определенных людей для обучения модели, она может "переобучиться" на особенностях именно этих людей, а не изучить общие закономерности (например, состояние человека: отправка сообщения, прием пищи и т.д.).

Аналогичная ситуация возникла в [конкурсе Kaggle по мониторингу рыболовства](https://www.kaggle.com/c/the-nature-conservancy-fisheries-monitoring), где целью было определение видов рыб, выловленных рыболовными судами, для борьбы с незаконным промыслом исчезающих популяций. В тестовом наборе были представлены суда, которые не встречались в обучающем наборе. Это означает, что в вашем наборе для проверки должны быть представлены суда, которых нет в обучающем наборе.

Иногда может быть неясно, чем именно будет отличаться ваш набор для проверки. Например, для задачи, использующей спутниковые снимки, необходимо собрать больше информации о том, содержал ли обучающий набор только снимки определенных географических регионов, или были ли они получены из данных, охватывающих различные географические области.
