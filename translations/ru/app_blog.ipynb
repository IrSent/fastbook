# Создание блога


К сожалению, когда дело доходит до ведения блогов, кажется, что вам приходится делать сложный выбор: либо использовать платформу, которая упрощает процесс, но подвергает вас и ваших читателей рекламе, платным подпискам и сборам, либо тратить часы на настройку собственного хостинга и недели на изучение множества сложных деталей. Возможно, самое большое преимущество подхода "сделай сам" заключается в том, что вы действительно владеете своими публикациями, а не зависите от поставщика услуг и его решений о том, как монетизировать ваш контент в будущем.

Однако, как оказалось, можно получить все преимущества обеих сторон!

## Ведение блога с помощью GitHub Pages


Отличным решением является размещение вашего блога на платформе [GitHub Pages](https://pages.github.com/), которая бесплатна, не содержит рекламы и платных функций, и делает ваши данные доступными в стандартном формате, что позволяет вам в любой момент перенести свой блог на другой хостинг. Однако все подходы к использованию GitHub Pages, которые мне встречались, требовали знания командной строки и сложных инструментов, с которыми, скорее всего, знакомы только разработчики программного обеспечения. Например, в [официальной документации](https://help.github.com/en/github/working-with-github-pages/creating-a-github-pages-site-with-jekyll) GitHub, посвященной настройке блога, приводится длинный список инструкций, включающий установку языка программирования Ruby, использование утилиты командной строки `git`, копирование номеров версий и многое другое — всего 17 шагов!

Чтобы упростить процесс, мы разработали простой способ, который позволяет вам использовать *полностью веб-интерфейс* для всех ваших задач, связанных с ведением блога. Вы сможете начать работу с вашим новым блогом примерно за пять минут. Это не требует никаких затрат, и вы можете легко добавить свой собственный домен, если захотите. В этом разделе мы объясним, как это сделать, используя шаблон, который мы создали и назвали *fast_template*. (Обратите внимание: обязательно посетите [сайт книги](https://book.fast.ai) для получения самых актуальных рекомендаций по ведению блога, так как постоянно появляются новые инструменты).

### Создание репозитория


Вам потребуется учетная запись на GitHub, поэтому перейдите туда и создайте ее, если у вас ее еще нет. Убедитесь, что вы вошли в систему. Обычно GitHub используется разработчиками программного обеспечения для написания кода, и они используют сложный инструмент командной строки для работы с ним, но мы покажем вам подход, который вообще не использует командную строку!

Чтобы начать, откройте в своем браузере страницу [https://github.com/fastai/fast_template/generate](https://github.com/fastai/fast_template/generate) (вам необходимо войти в систему GitHub, чтобы ссылка работала). Это позволит вам создать место для хранения вашего блога, которое называется *репозиторием*. Вы увидите экран, похожий на тот, что показан на рисунке <<github_repo>>. Обратите внимание, что вам нужно ввести имя репозитория в *точном* формате, указанном здесь: ваше имя пользователя GitHub, за которым следует `.github.io`.

<img width="440" src="images/fast_template/image1.png" id="github_repo" caption="Создание вашего репозитория" alt="Скриншот страницы GitHub для создания нового репозитория">

После того, как вы введете имя и, при желании, описание, нажмите кнопку "Создать репозиторий из шаблона". У вас есть возможность сделать репозиторий "частным", но поскольку вы создаете блог, который хотите, чтобы другие люди читали, наличие общедоступных исходных файлов, скорее всего, не станет для вас проблемой.

Теперь давайте настроим вашу главную страницу!


### Настройка главной страницы


Когда читатели заходят на ваш блог, первое, что они видят, это содержимое файла с именем *index.md*. Это файл, написанный на языке [Markdown](https://guides.github.com/features/mastering-markdown/). Markdown – это мощный, но простой способ создания форматированного текста, например, списков, курсива, гиперссылок и так далее. Он широко используется, в том числе для форматирования в блокнотах Jupyter, почти во всех разделах сайта GitHub и во многих других местах в интернете. Чтобы создать текст на языке Markdown, вы можете просто набирать текст на английском языке, а затем добавлять специальные символы для придания тексту определенных свойств. Например, если вы поставите символ `*` перед и после слова или фразы, оно будет выделено *курсивом*. Давайте попробуем это сейчас.

Чтобы открыть файл, нажмите на его имя на сайте GitHub. Чтобы отредактировать его, нажмите на значок карандаша, расположенный в правом верхнем углу экрана, как показано на рисунке <<fastpage_edit>>.

<img width="800" src="images/fast_template/image3.png" alt="Скриншот, показывающий, куда нажать для редактирования файла" id="fastpage_edit" caption="Нажмите «Редактировать этот файл»">

Вы можете добавлять, редактировать или заменять текст, который видите. Нажмите кнопку "Предварительный просмотр" (<<fastpage_preview>>) чтобы увидеть, как будет выглядеть ваш текст в формате Markdown на вашем блоге. Строки, которые вы добавили или изменили, будут отображаться с зеленой полосой слева.

<img width="350" src="images/fast_template/image4.png" alt="Скриншот, показывающий, куда нажать для предварительного просмотра изменений" id="fastpage_preview" caption="Предварительный просмотр изменений, чтобы выявить возможные ошибки">

Чтобы сохранить изменения, прокрутите страницу вниз и нажмите кнопку "Сохранить изменения" (как показано на <<fastpage_commit>>). В GitHub "сохранить изменения" означает сохранить их на сервере GitHub.

<img width="600" src="images/fast_template/image5.png" alt="Скриншот, показывающий, куда нажать для сохранения изменений" id="fastpage_commit" caption="Сохраните изменения, чтобы они были применены">

Далее необходимо настроить параметры вашего блога. Для этого откройте файл с именем *\_config.yml*, а затем нажмите кнопку "Редактировать", как вы делали это для файла index. Измените значения заголовка, описания и имени пользователя GitHub (см. рисунок <<github_config>>). Вам нужно оставить названия, указанные перед двоеточием, и ввести ваши новые значения после двоеточия (и пробела) на каждой строке. Вы также можете добавить свой адрес электронной почты и имя пользователя Twitter, если хотите, но имейте в виду, что эта информация будет отображаться на вашем публичном блоге, если вы ее укажете.

<img width="800" src="images/fast_template/image6.png" id="github_config" caption="Заполните файл конфигурации" alt="Скриншот, показывающий файл конфигурации и способ его заполнения">

После завершения внесите изменения и зафиксируйте их, как вы делали это с файлом index, затем подождите около минуты, пока GitHub обработает ваш новый блог. Откройте в своем веб-браузере адрес *&lt;username> .github.io* (заменив *&lt;username>* на ваше имя пользователя GitHub). Вы должны увидеть свой блог, который будет выглядеть примерно так, как на рисунке <<github_blog>>.

<img width="540" src="images/fast_template/image7.png" id="github_blog" caption="Ваш блог онлайн!" alt="Скриншот веб-сайта username.github.io">

### Создание публикаций


Теперь вы готовы создать свою первую публикацию. Все ваши публикации будут храниться в папке *\_posts*. Нажмите на нее, а затем нажмите кнопку "Создать файл". Важно назвать ваш файл в формате *&lt;год>-&lt;месяц>-&lt;день>-&lt;название>.md*, как показано на рисунке <<fastpage_name>>, где *&lt;год>* – это четырехзначное число, а *&lt;месяц>* и *&lt;день>* – двузначные числа. *&lt;название>* может быть любым, что поможет вам вспомнить, о чем была эта публикация. Расширение *.md* предназначено для документов, написанных на языке Markdown.

<img width="440" src="images/fast_template/image8.png" alt="Скриншот, показывающий правильный синтаксис для создания новой публикации в блоге" id="fastpage_name" caption="Название ваших публикаций">

Затем вы можете ввести содержимое вашей первой публикации. Единственное правило: первая строка вашей публикации должна быть заголовком, написанным на языке Markdown. Это создается путем добавления символа `# ` в начале строки, как показано на рисунке <<fastpage_title>> (это создает заголовок первого уровня, который следует использовать только один раз в начале вашего документа; вы можете создавать заголовки второго уровня, используя `## `, третьего уровня – `###` и так далее).

<img width="300" src="images/fast_template/image9.png" alt="Скриншот, показывающий начало публикации в блоге" id="fastpage_title" caption="Синтаксис Markdown для заголовка">

Как и раньше, вы можете нажать кнопку "Предварительный просмотр", чтобы увидеть, как будет выглядеть форматирование вашего текста с использованием Markdown (<<fastpage_preview1>>).

<img width="400" src="images/fast_template/image10.png" alt="Скриншот, демонстрирующий, как тот же самый пост в блоге будет отображаться в HTML" id="fastpage_preview1" caption="Как будет выглядеть предыдущий синтаксис Markdown на вашем блоге">

Затем вам нужно будет нажать кнопку "Сохранить новый файл", чтобы сохранить его на GitHub, как показано на рисунке <<fastpage_commit1>>.

<img width="700" src="images/fast_template/image11.png" alt="Скриншот, показывающий, куда нажать для сохранения нового файла" id="fastpage_commit1" caption="Сохраните изменения, чтобы они вступили в силу">

Посмотрите еще раз на главную страницу вашего блога, и вы увидите, что эта запись теперь появилась. <<fastpage_live>> показывает результат с использованием добавленной нами демонстрационной фотографии. (Помните, что вам может потребоваться подождать минуту или около того, пока GitHub обработает запрос, прежде чем файл появится.)

<img width="500" src="images/fast_template/image12.png" alt="Скриншот, показывающий первую запись на сайте блога" id="fastpage_live" caption="Ваша первая запись опубликована!">

Возможно, вы заметили, что мы предоставили демонстрационную запись для блога, которую вы можете сейчас удалить. Перейдите в папку *\_posts*, как и раньше, и щелкните по файлу *2020-01-14-welcome.md*. Затем нажмите на значок корзины в правом верхнем углу, как показано на <<fastpage_delete>>.

<img width="400" src="images/fast_template/image13.png" alt="Скриншот, показывающий, как удалить демонстрационную запись" id="fastpage_delete" caption="Удалите демонстрационную запись для блога">

В GitHub ничего не меняется, пока вы не зафиксируете изменения, включая удаление файлов! Поэтому, после того как вы нажмете на значок корзины, прокрутите страницу вниз и зафиксируйте внесенные изменения.

Вы можете добавлять изображения в свои публикации, используя строчку в формате Markdown, например:

    ![Описание изображения](images/имя_файла.jpg)

Чтобы это работало, вам нужно поместить изображение в папку *images*. Для этого нажмите на папку *images*, затем нажмите кнопку "Загрузить файлы" (<<fastpage_upload>>).

<img width="400" src="images/fast_template/image14.png" alt="Скриншот, показывающий, как загрузить новые файлы" id="fastpage_upload" caption="Загрузить файл с вашего компьютера">

Теперь давайте посмотрим, как выполнить все это непосредственно с вашего компьютера.

### Синхронизация GitHub и вашего компьютера


Существует множество причин, по которым вам может потребоваться скопировать контент вашего блога с GitHub на ваш компьютер: возможно, вы хотите иметь возможность читать или редактировать свои записи в автономном режиме, или, возможно, вам нужна резервная копия на случай, если что-то произойдет с вашим репозиторием на GitHub.

GitHub позволяет не только копировать ваш репозиторий на ваш компьютер, но и *синхронизировать* его с вашим компьютером. Это означает, что вы можете вносить изменения на GitHub, и они будут автоматически копироваться на ваш компьютер, а также вы можете вносить изменения на вашем компьютере, и они будут автоматически копироваться на GitHub. Вы даже можете предоставить другим людям доступ к вашему блогу и разрешить им вносить изменения, и все изменения, внесенные вами и другими, будут автоматически объединены при следующей синхронизации.

Чтобы это работало, вам необходимо установить приложение под названием [GitHub Desktop](https://desktop.github.com/) на ваш компьютер. Оно работает на Mac, Windows и Linux. Следуйте инструкциям по установке, и при запуске приложения вам будет предложено войти в учетную запись GitHub и выбрать репозиторий для синхронизации. Нажмите "Клонировать репозиторий из интернета", как показано на рисунке <<fastpage_clone>>.

<img src="images/gitblog/image1.png" width="400" alt="Скриншот, показывающий, как клонировать ваш репозиторий" id="fastpage_clone" caption="Клонируйте свой репозиторий в GitHub Desktop">

Как только GitHub завершит синхронизацию вашего репозитория, вы сможете нажать кнопку "Просмотреть файлы вашего репозитория в проводнике" (или Finder), как показано на рисунке <<fastpage_explorer>>, и увидите локальную копию вашего блога! Попробуйте отредактировать один из файлов на своем компьютере. Затем вернитесь в GitHub Desktop, и вы увидите кнопку "Синхронизировать", которая ждет, пока вы ее нажмете. После нажатия этой кнопки ваши изменения будут скопированы в GitHub, где вы увидите их отражение на веб-сайте.

<img src="images/gitblog/image2.png" width="600" alt="Скриншот клонированного репозитория" id="fastpage_explorer" caption="Просмотр файлов локально">

Если вы раньше не пользовались `git`, GitHub Desktop – отличный способ начать. Как вы убедитесь, это важный инструмент, который используют большинство специалистов по работе с данными. Еще один инструмент, который, мы надеемся, вам понравится, – это Jupyter Notebooks, и есть возможность даже писать свои статьи в блог непосредственно в нем!

## Jupyter для ведения блогов.


Вы также можете создавать записи для блога, используя Jupyter Notebook. Ваши ячейки с текстом в формате Markdown, ячейки с кодом и все результаты выполнения будут отображаться в вашей экспортированной записи для блога. Возможно, лучший способ сделать это изменился к моменту, когда вы читаете эту книгу, поэтому обязательно посетите [сайт книги](https://book.fast.ai), чтобы узнать самую актуальную информацию. На данный момент самый простой способ создать блог из блокнотов — использовать [`fastpages`](http://fastpages.fast.ai/), который является более продвинутой версией `fast_template`.

Чтобы добавить блокнот в блог, просто поместите его в папку *\_notebooks* в репозитории вашего блога, и он появится в списке ваших записей. Когда вы пишете блокнот, пишите все, что вы хотите, чтобы ваша аудитория увидела. Поскольку большинство платформ для написания затрудняют включение кода и результатов выполнения, многие из нас привыкли включать меньше реальных примеров, чем следовало бы. Это отличный способ выработать привычку включать много примеров во время написания.

Часто вам захочется скрыть шаблонный код, например, операторы импорта. Вы можете добавить `#hide` в начало любой ячейки, чтобы она не отображалась в результатах. Jupyter отображает результат последней строки в ячейке, поэтому нет необходимости использовать `print()`. (Включение лишнего кода, который не нужен, создает дополнительную нагрузку на читателя, поэтому не включайте код, который вам действительно не нужен!)
