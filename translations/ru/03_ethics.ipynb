### Конец боковой панели


Как мы обсуждали в главах 1 и 2, иногда модели машинного обучения могут работать некорректно. В них могут быть ошибки. Они могут получать данные, которых раньше не видели, и вести себя непредсказуемым образом. Или они могут работать именно так, как задумано, но использоваться для целей, для которых мы бы предпочли, чтобы их никогда не использовали.

Поскольку глубокое обучение – это мощный инструмент, который можно использовать для множества задач, особенно важно, чтобы мы учитывали последствия наших решений. Философское изучение *этики* – это изучение добра и зла, включая то, как мы можем определить эти понятия, распознавать правильные и неправильные действия и понимать связь между действиями и их последствиями. Область *этики данных* существует уже давно, и многие ученые занимаются исследованиями в этой области. Она используется для разработки политик во многих юрисдикциях, в крупных и малых компаниях для рассмотрения того, как лучше обеспечить положительные социальные результаты при разработке продуктов, а также исследователями, которые хотят убедиться, что их работа используется во благо, а не во вред.

Следовательно, как специалисту в области глубокого обучения, вероятно, вы окажетесь в ситуации, когда вам придется учитывать этические аспекты работы с данными. Итак, что такое этика данных? Это подраздел этики, поэтому давайте начнем с этого.

J: В университете я в основном занимался философией этики (это была бы тема моей диссертации, если бы я ее закончил, а не бросил, чтобы "вернуться в реальный мир"). Основываясь на годах, которые я провел, изучая этику, я могу сказать вам следующее: никто по-настоящему не согласен в том, что такое правильно и неправильно, существуют ли они, как их распознать, какие люди хорошие, а какие плохие, или практически в чем-либо еще. Так что не стоит ожидать слишком многого от теории! Мы сосредоточимся на примерах и вопросах для размышления, а не на теории.

В ответе на вопрос ["Что такое этика?"](https://www.scu.edu/ethics/ethics-resources/ethical-decision-making/what-is-ethics/), Центр прикладной этики Маркулы определяет этот термин следующим образом:

- Это четко сформулированные принципы добра и зла, которые определяют, что люди должны делать.
- Это изучение и развитие собственных этических принципов.

Не существует списка правильных ответов. Не существует списка того, что можно и что нельзя делать. Этика – это сложно и зависит от контекста. Она включает в себя точки зрения различных заинтересованных сторон. Этика – это навык, который необходимо развивать и тренировать. В этой главе наша цель – предоставить вам некоторые ориентиры, которые помогут вам в этом процессе.

Выявление этических проблем лучше всего проводить в рамках совместной работы. Это единственный способ по-настоящему учесть различные точки зрения. Разный жизненный опыт поможет людям увидеть то, что может быть не очевидным для вас. Работа в команде полезна для многих задач, направленных на развитие навыков, включая и эту.

Эта глава, безусловно, не является единственной частью книги, где мы говорим об этике данных, но полезно иметь место, где мы можем сосредоточиться на этом вопросе. Чтобы сориентироваться, возможно, проще всего рассмотреть несколько примеров. Поэтому мы выбрали три примера, которые, как нам кажется, хорошо иллюстрируют некоторые ключевые темы.

### Ошибки и способы их исправления: Неправильный алгоритм, используемый для расчета пособий по медицинскому страхованию.

Сайт The Verge провел расследование программного обеспечения, используемого более чем в половине штатов США для определения объема медицинской помощи, предоставляемой гражданам, и представил результаты своего исследования в статье ["Что происходит, когда алгоритм сокращает объем вашей медицинской помощи"](https://www.theverge.com/2018/3/21/17144260/healthcare-medicaid-algorithm-arkansas-cerebral-palsy). После внедрения алгоритма в штате Арканзас сотни людей (многие из которых имеют серьезные нарушения здоровья) столкнулись со значительным сокращением объема предоставляемой им медицинской помощи. Например, Тэмми Доббс, женщина, страдающая церебральным параличом и нуждающаяся в помощи для вставания с постели, посещения туалета, приема пищи и т.д., внезапно обнаружила, что объем оказываемой ей помощи сократился на 20 часов в неделю. Она не получила никаких объяснений причин такого сокращения. В результате судебного разбирательства выяснилось, что в процессе внедрения алгоритма были допущены ошибки, которые негативно повлияли на людей, страдающих диабетом или церебральным параличом. Однако Тэмми Доббс и многие другие люди, зависящие от этих видов медицинской помощи, живут в страхе, что объем их помощи может снова быть внезапно и необъяснимо сокращен.

### Обратная связь: Система рекомендаций YouTube


Обратные связи могут возникать, когда ваша модель контролирует следующий пакет данных, который вы получаете. Данные, которые возвращаются, очень быстро становятся неточными из-за самого программного обеспечения.

Например, YouTube имеет 1,9 миллиарда пользователей, которые ежедневно смотрят более 1 миллиарда часов видео. Алгоритм рекомендаций (разработанный компанией Google), который был создан для оптимизации времени просмотра, отвечает примерно за 70% контента, который пользователи смотрят. Однако возникла проблема: он привел к неконтролируемым обратным связям, что побудило газету *New York Times* опубликовать статью под заголовком ["YouTube спровоцировал бум теорий заговора. Можно ли это остановить?"](https://www.nytimes.com/2019/02/19/technology/youtube-conspiracy-stars.html). По сути, системы рекомендаций предсказывают, какой контент понравится пользователям, но они также обладают огромной властью в определении того, какой контент пользователи вообще видят.

### Предвзятость: Профессор Латания Суини "задержана"


Доктор Латания Суини – профессор Гарвардского университета и директор лаборатории по вопросам защиты данных этого университета. В статье ["Дискриминация в таргетированной онлайн-рекламе"](https://arxiv.org/abs/1301.6822) (см. <<latanya_arrested>>) она описывает свое открытие: при поиске в Google по ее имени появлялась реклама с надписью "Латания Суини, арестована?", хотя она является единственной известной личностью с таким именем и никогда не была арестована. Однако, когда она искала в Google другие имена, например, "Кирстен Линдквист", она получала более нейтральную рекламу, хотя Кирстен Линдквист была арестована трижды.

Это пример предвзятости. Это может существенно повлиять на жизнь людей: например, если информацию о кандидате на работу ищут в Google, может создаться впечатление, что у него есть судимость, даже если это не так.

### Почему это важно?


Одна из самых естественных реакций при рассмотрении этих вопросов: "И что? Что это вообще имеет ко мне отношение? Я специалист по анализу данных, а не политик. Я не один из руководителей моей компании, которые принимают решения о том, что мы делаем. Я просто пытаюсь создать наиболее точную модель, насколько это возможно."

Это вполне разумные вопросы. Но мы постараемся убедить вас, что ответ заключается в том, что каждый, кто разрабатывает модели, должен учитывать, как эти модели будут использоваться, и подумать о том, как лучше всего обеспечить их максимально позитивное применение. Есть вещи, которые вы можете сделать. И если вы этого не сделаете, последствия могут быть очень серьезными.

Один особенно отвратительный пример того, что происходит, когда технологи сосредотачиваются исключительно на технологиях, – это история IBM и нацистской Германии. В 2001 году швейцарский судья постановил, что "не было бы необоснованным предположить, что техническая помощь IBM способствовала задачам, которые нацисты выполняли при совершении преступлений против человечества, в том числе с использованием машин IBM для учета и классификации, которые использовались непосредственно в концентрационных лагерях".

Как видите, компания IBM поставляла нацистам продукты для обработки данных, необходимые для отслеживания массового уничтожения евреев и других групп. Это было инициировано руководством компании, в том числе путем маркетинга продукции Гитлеру и его команде. Президент компании Томас Уотсон лично одобрил выпуск в 1939 году специальных машин для сортировки данных, которые использовались для организации депортации польских евреев. На фотографии, расположенной ниже, изображен Адольф Гитлер (слева) во время встречи с генеральным директором IBM Томом Уотсоном-старшим (второй слева) незадолго до того, как Гитлер вручил Уотсону специальную медаль "За заслуги перед Рейхом" в 1937 году.

Подумайте на мгновение: что бы вы почувствовали, если бы узнали, что вы были частью системы, которая в итоге нанесла вред обществу? Были бы вы готовы узнать правду? Как вы можете помочь предотвратить это? Мы описали здесь самую крайнюю ситуацию, но существует множество негативных социальных последствий, связанных с искусственным интеллектом и машинным обучением, которые наблюдаются сегодня, и некоторые из них мы рассмотрим в этой главе.

Это не только моральная ответственность. Иногда специалисты в области технологий напрямую расплачиваются за свои действия. Например, первым человеком, попавшим в тюрьму в результате скандала с Volkswagen, когда компания была уличена в фальсификации результатов тестов выбросов дизельных двигателей, был не руководитель проекта или топ-менеджер компании. Это был один из инженеров, Джеймс Лян, который просто выполнял то, что ему было сказано.

Конечно, не все так плохо: если проект, в котором вы участвуете, оказывает огромное положительное влияние даже на одного человека, это принесет вам огромное удовлетворение!

Итак, надеюсь, мы убедили вас в том, что вам следует проявлять заинтересованность. Но что вы должны делать? Как специалисты по анализу данных, мы, как правило, стремимся улучшить наши модели, оптимизируя определенные показатели. Однако оптимизация этих показателей не всегда приводит к лучшим результатам. И даже если это *действительно* помогает добиться лучших результатов, это почти наверняка не единственный важный фактор. Рассмотрите цепочку этапов, которая проходит между разработкой модели или алгоритма исследователем или специалистом, и моментом, когда эта работа фактически используется для принятия какого-либо решения. Необходимо рассматривать всю эту цепочку *в целом*, если мы хотим добиться желаемых результатов.

Обычно это очень длинная цепочка от начала до конца. Это особенно верно, если вы исследователь, и вы можете даже не знать, будет ли ваше исследование когда-либо использовано для чего-либо, или если вы участвуете в сборе данных, что является еще более ранним этапом. Но никто лучше вас не может информировать всех, кто участвует в этой цепочке, о возможностях, ограничениях и деталях вашей работы. Хотя нет "волшебной таблетки", которая могла бы гарантировать правильное использование вашей работы, участвуя в этом процессе и задавая правильные вопросы, вы, по крайней мере, можете обеспечить рассмотрение важных вопросов.

Иногда правильный ответ на просьбу выполнить определенную работу — это просто сказать "нет". Часто, однако, мы слышим ответ: "Если я этого не сделаю, это сделает кто-то другой". Но подумайте об этом: если вас выбрали для этой работы, вы – лучший человек, которого они нашли для ее выполнения, – значит, если вы этого не сделаете, лучший специалист не будет работать над этим проектом. Если первые пять человек, которых они попросили, тоже отказываются, это еще лучше!


## Интеграция машинного обучения с разработкой продуктов.


Предположительно, вы выполняете эту работу, потому что надеетесь, что она будет использована для чего-то. В противном случае, вы просто тратите свое время. Итак, давайте исходим из предположения, что ваша работа где-то окажется. Теперь, когда вы собираете данные и разрабатываете свою модель, вы принимаете множество решений. На каком уровне детализации вы будете хранить свои данные? Какую функцию потерь следует использовать? Какие наборы данных для валидации и обучения вы будете использовать? Стоит ли вам сосредоточиться на простоте реализации, скорости работы или точности модели? Как ваша модель будет обрабатывать данные, выходящие за рамки ее области применения? Можно ли ее дообучить, или ее необходимо будет переобучать с нуля со временем?

Это не просто вопросы, касающиеся алгоритмов. Это вопросы проектирования продуктов, основанных на данных. Но менеджеры продуктов, руководители, судьи, журналисты, врачи... все, кто в конечном итоге будет разрабатывать и использовать систему, частью которой является ваша модель, не смогут понять решения, которые вы приняли, не говоря уже о том, чтобы изменить их.

Например, два исследования показали, что программное обеспечение для распознавания лиц, разработанное компанией Amazon, выдавало [неточные](https://www.nytimes.com/2018/07/26/technology/amazon-aclu-facial-recognition-congress.html) и [расово предвзятые](https://www.theverge.com/2019/1/25/18197137/amazon-rekognition-facial-recognition-bias-race-gender) результаты. Компания Amazon заявила, что исследователи должны были изменить параметры по умолчанию, не объяснив, как это могло бы исправить предвзятые результаты. Кроме того, выяснилось, что [Amazon не давала указаний полицейским управлениям](https://gizmodo.com/defense-of-amazons-face-recognition-tool-undermined-by-1832238149), использующим ее программное обеспечение для этих целей. Вероятно, существовала большая разница между исследователями, разрабатывавшими эти алгоритмы, и сотрудниками Amazon, занимавшимися документацией и составлявшими руководства для полиции. Отсутствие тесной интеграции привело к серьезным проблемам для общества в целом, для полиции и для самой компании Amazon. Оказалось, что их система ошибочно сопоставила 28 членов конгресса с фотографиями, сделанными при задержании! (И среди членов конгресса, ошибочно сопоставленных с фотографиями при задержании, непропорционально много были представители расовых меньшинств, как показано в документе <<congressmen>>.)

Специалисты по анализу данных должны быть частью междисциплинарной команды. Исследователям необходимо тесно сотрудничать с теми людьми, которые в конечном итоге будут использовать результаты их исследований. Было бы еще лучше, если бы сами эксперты в своей области приобрели достаточно знаний, чтобы самостоятельно обучать и отлаживать некоторые модели. Надеемся, что среди читателей этой книги есть такие люди!

Современное рабочее место – это очень специализированная среда. Обычно у каждого сотрудника есть четко определенные задачи. Особенно в крупных компаниях бывает трудно понять, какие именно элементы составляют общую картину. Иногда компании даже намеренно скрывают общие цели проектов, над которыми работают, если они знают, что их сотрудникам не понравятся ответы. Это часто делается путем максимального разделения задач.

Другими словами, мы не говорим, что все это легко. Это сложно. Очень сложно. Мы все должны приложить максимум усилий. Мы часто видим, что люди, которые вовлекаются в более широкий контекст этих проектов и стремятся развивать междисциплинарные навыки и команды, становятся одними из самых важных и ценных членов своих организаций. Это работа, которая, как правило, высоко ценится высшим руководством, даже если она иногда кажется неудобной для среднего звена управления.

## Вопросы этики в области обработки данных.


Этика данных – это обширная область, и мы не можем охватить все аспекты. Вместо этого, мы выберем несколько тем, которые, по нашему мнению, особенно важны:

- Необходимость механизмов обжалования и привлечения к ответственности.
- Обратная связь.
- Предвзятость.
- Дезинформация.

Рассмотрим каждый из них по отдельности.

### Возможности для обжалования и привлечения к ответственности.


В сложной системе легко, чтобы никто не чувствовал себя ответственным за результаты. Хотя это и понятно, это не приводит к хорошим результатам. В приведенном ранее примере системы здравоохранения штата Арканзас, где из-за ошибки люди с церебральным параличом потеряли доступ к необходимой медицинской помощи, создатель алгоритма обвинил государственных чиновников, а государственные чиновники – тех, кто внедрил программное обеспечение. Профессор Нью-Йоркского университета [Дана Бойд](https://www.youtube.com/watch?v=NTl0yyPqf3E) описала это явление следующим образом: "Бюрократия часто используется для перекладывания или уклонения от ответственности... Современные алгоритмические системы расширяют бюрократию."

Еще одна причина, по которой так необходимы механизмы контроля и исправления ошибок, заключается в том, что данные часто содержат ошибки. Механизмы аудита и исправления ошибок имеют решающее значение. Была обнаружена, что база данных, содержащая информацию о предполагаемых членах банд, которую ведут правоохранительные органы Калифорнии, была полна ошибок, включая данные о 42 младенцах, которые были внесены в базу данных в возрасте менее одного года (28 из которых были отмечены как "признающие себя членами банды"). В этом случае не было предусмотрено никаких процедур для исправления ошибок или удаления людей после того, как они были внесены в базу данных. Другой пример – система кредитных отчетов в США: в масштабном исследовании кредитных отчетов, проведенном Федеральной торговой комиссией (FTC) в 2012 году, было установлено, что у 26% потребителей есть хотя бы одна ошибка в их файлах, а у 5% – ошибки, которые могут иметь разрушительные последствия. Однако процесс исправления таких ошибок невероятно медленный и непрозрачный. Когда журналист радиостанции [Бобби Эллин](https://www.washingtonpost.com/posteverything/wp/2016/09/08/how-the-careless-errors-of-credit-reporting-agencies-are-ruining-peoples-lives/) обнаружил, что он ошибочно был внесен в список лиц, имеющих судимость за хранение оружия, ему потребовалось "более дюжины телефонных звонков, помощь сотрудника окружного суда и шесть недель, чтобы решить проблему. И это было только после того, как я обратился в отдел коммуникаций компании как журналист".

Как специалисты в области машинного обучения, мы не всегда считаем своей обязанностью понимать, как наши алгоритмы применяются на практике. Но нам нужно это понимать.

### Обратная связь


В разделе <<chapter_intro>> мы объяснили, как алгоритм может взаимодействовать со своей средой, создавая петлю обратной связи, генерируя прогнозы, которые усиливают действия, совершаемые в реальном мире, что приводит к прогнозам, еще более выраженным в том же направлении.
В качестве примера, давайте снова рассмотрим систему рекомендаций YouTube. Несколько лет назад команда Google рассказала о том, как они внедрили обучение с подкреплением (тесно связанное с глубоким обучением, но где функция потерь представляет собой результат, который может наступить спустя долгое время после совершения действия), чтобы улучшить систему рекомендаций YouTube. Они описали, как использовали алгоритм, который выдавал рекомендации таким образом, чтобы оптимизировать время просмотра.

Однако, люди, как правило, тянутся к спорному контенту. Это означало, что видеоролики, посвященные таким темам, как теории заговора, стали все чаще рекомендоваться системой. Кроме того, оказалось, что люди, интересующиеся теориями заговора, также активно смотрят онлайн-видео! Поэтому они все больше и больше стали привлекаться к YouTube. Увеличение числа сторонников теорий заговора, смотрящих видео на YouTube, привело к тому, что алгоритм стал рекомендовать все больше и больше контента, посвященного теориям заговора и другим экстремистским материалам, что, в свою очередь, привело к тому, что еще больше экстремистов стали смотреть видео на YouTube, и что все больше людей, смотрящих YouTube, начали придерживаться экстремистских взглядов, что, в конечном итоге, привело к тому, что алгоритм стал рекомендовать еще больше экстремистского контента... Система вышла из-под контроля.

И это явление не ограничивалось только этим типом контента. В июне 2019 года газета *New York Times* опубликовала статью о системе рекомендаций YouTube под названием ["На цифровой площадке YouTube открыт проход для педофилов"](https://www.nytimes.com/2019/06/03/world/americas/youtube-pedophiles.html). Статья начиналась с этой тревожной истории:


: Кристиан C. не придала особого значения, когда ее 10-летняя дочь и ее подруга загрузили видео, на котором они играют в бассейне во дворе… Через несколько дней… видео набрало тысячи просмотров. Вскоре количество просмотров достигло 400 000... "Я снова посмотрела это видео и испугалась от количества просмотров", - сказала Кристиан. У нее были веские основания для этого. Автоматизированная система рекомендаций YouTube… начала показывать это видео пользователям, которые смотрели другие видео с участием детей в предпубертатном возрасте, частично раздевшихся, как установили исследователи.

: Само по себе каждое видео может быть совершенно безобидным, например, домашним фильмом, снятым ребенком. Любые кадры, содержащие что-либо, что может вызвать вопросы, являются мимолетными и кажутся случайными. Однако, собранные вместе, их общие черты становятся очевидными.

Алгоритм рекомендаций YouTube начал формировать плейлисты для педофилов, выбирая невинные домашние видео, в которых случайно оказывались дети в предпубертатном возрасте, частично одетые.

Никто в Google не планировал создавать систему, которая превращала бы семейные видео в порнографию для педофилов. Так что же произошло?

Часть проблемы заключается в том, что ключевую роль в работе финансово важной системы играют показатели. Как вы уже видели, когда у алгоритма есть показатель, который необходимо оптимизировать, он будет делать все возможное, чтобы оптимизировать это число. Это часто приводит к возникновению самых разных ситуаций, и люди, взаимодействующие с системой, будут искать, находить и использовать эти "крайние случаи" и обратные связи для своей выгоды.

Существуют признаки того, что именно это и произошло с системой рекомендаций YouTube. Издание *The Guardian* опубликовало статью под названием ["Как бывший сотрудник YouTube расследовал его секретный алгоритм"](https://www.theguardian.com/technology/2018/feb/02/youtube-algorithm-election-clinton-trump-guillaume-chaslot), посвященную Гийому Шасло, бывшему инженеру YouTube, который создал AlgoTransparency, отслеживающий эти проблемы. Шасло опубликовал диаграмму в документе <<ethics_yt_rt>>, после публикации отчета Роберта Мюллера о расследовании вмешательства России в президентские выборы 2016 года.

<img src="images/ethics/image18.jpeg" id="ethics_yt_rt" caption="Освещение отчета Мюллера" alt="Освещение отчета Мюллера" width="500">

Освещение доклада Мюллера на телеканале Russia Today значительно отличалось от того, как его рекомендовали другие каналы. Это указывает на возможность того, что Russia Today, российское государственное СМИ, успешно манипулирует алгоритмом рекомендаций YouTube. К сожалению, отсутствие прозрачности в работе подобных систем затрудняет выявление проблем, о которых мы говорим.

Один из наших рецензентов этой книги, Орельен Жерон, руководил командой по классификации видео в YouTube с 2013 по 2016 год (задолго до событий, описанных здесь). Он отметил, что проблема заключается не только в обратных связях, в которых участвуют люди. Могут существовать обратные связи и без участия людей! Он рассказал нам пример из YouTube:

> : Одним из важных факторов для определения основной темы видео является канал, на котором оно размещено. Например, видео, загруженное на кулинарный канал, скорее всего, будет посвящено кулинарии. Но как мы можем узнать, о чем канал? Ну, частично, исходя из тем видео, которые на нем размещены! Видите ли вы эту петлю? Например, многие видео содержат описание, в котором указано, какая камера использовалась для съемки. В результате, некоторые из этих видео могут быть классифицированы как видео о "фотографии". Если канал содержит видео, которое ошибочно классифицировано таким образом, он может быть классифицирован как канал о "фотографии", что еще больше повышает вероятность того, что будущие видео на этом канале будут неправильно классифицированы как "фотография". Это может привести к цепной реакции неконтролируемой классификации! Один из способов разорвать эту петлю обратной связи — классифицировать видео с использованием и без информации о канале. Затем, при классификации каналов, можно использовать только те категории, которые были получены без информации о канале. Таким образом, петля обратной связи прерывается.

Существуют положительные примеры людей и организаций, пытающихся бороться с этими проблемами. Эван Эстола, ведущий инженер по машинному обучению в Meetup, [обсудил пример](https://www.youtube.com/watch?v=MqoRzNhrTnQ) того, как мужчины проявляют больший интерес к технологическим встречам, чем женщины. Учет гендерных факторов может привести к тому, что алгоритм Meetup будет рекомендовать женщинам меньше технологических встреч, и, как следствие, меньше женщин узнает о них и посетит их, что, в свою очередь, может привести к тому, что алгоритм будет предлагать женщинам еще меньше технологических встреч, и так далее, создавая самоподдерживающуюся петлю обратной связи. Поэтому Эван и его команда приняли этичное решение не допускать создания такой петли обратной связи в своем алгоритме рекомендаций, явно не используя гендерные данные в этой части своей модели. Приятно видеть, как компания не просто слепо оптимизирует какой-либо показатель, но и учитывает его влияние. По словам Эвана, "Вам нужно решить, какие параметры не использовать в своем алгоритме... возможно, самый оптимальный алгоритм – это не лучший для внедрения в производство".

В то время как Meetup решила избежать такого исхода, Facebook является примером того, как можно позволить неконтролируемой петле обратной связи развиваться. Как и YouTube, она имеет тенденцию радикализовать пользователей, интересующихся одной теорией заговора, знакомя их с другими. Как пишет Рене Диреста, исследователь, занимающаяся распространением дезинформации, [в своей статье](https://www.fastcompany.com/3059742/social-network-algorithms-are-distorting-reality-by-boosting-conspiracy-theories):


> Как только люди присоединяются к одной группе в [Facebook], которая пропагандирует теории заговора, алгоритмы автоматически направляют их в множество других подобных групп. Если вы присоединяетесь к группе, выступающей против вакцинации, вам будут предлагаться группы, посвященные другим темам, таким как анти-ГМО, теории о химических следах в небе, сторонники теории плоской Земли (да, действительно) и группы, пропагандирующие "естественные методы лечения рака". Вместо того, чтобы вытащить пользователя из "кроличьей норы" (т.е. из мира теорий заговора), система рекомендаций, наоборот, подталкивает его еще глубже в нее.

Крайне важно помнить, что подобное поведение может возникнуть, и либо предвидеть возникновение обратной связи, либо предпринять активные действия для ее прерывания, когда вы замечаете первые признаки этого в своих собственных проектах. Еще один важный момент, о котором стоит помнить, это *предвзятость*, которая, как мы кратко обсуждали в предыдущей главе, может взаимодействовать с обратными связями очень неблагоприятным образом.

### Предвзятость


Обсуждения предвзятости в интернете часто быстро становятся довольно запутанными. Слово "предвзятость" имеет очень много разных значений. Статистики часто считают, что когда специалисты по этике данных говорят о предвзятости, они имеют в виду статистическое определение этого термина. Но это не так. И, конечно, они не говорят о предвзятостях, которые проявляются в весах и смещениях, которые являются параметрами вашей модели!

Они говорят о концепции предвзятости, используемой в социальных науках. В работе ["A Framework for Understanding Unintended Consequences of Machine Learning"](https://arxiv.org/abs/1901.10002) Харини Суреш и Джон Гуттаг из Массачусетского технологического института описывают шесть типов предвзятости в машинном обучении, которые обобщены в разделе <<bias>> их работы.

Мы обсудим четыре типа таких предвзятостей, которые оказались наиболее полезными в нашей собственной работе (подробности о других типах можно найти в статье).

#### Историческая предвзятость.

*Историческая предвзятость* возникает из-за того, что люди предвзяты, процессы предвзяты, и общество предвзято. Суреш и Гуттаг утверждают: "Историческая предвзятость – это фундаментальная, структурная проблема, затрагивающая первый этап процесса сбора данных, и она может существовать даже при идеальной выборке и отборе признаков."

Вот несколько примеров исторической *расовой предвзятости* в США, взятых из статьи "Расовая предвзятость, даже когда у нас самые лучшие намерения" ("Racial Bias, Even When We Have Good Intentions"), опубликованной в *The New York Times* и написанной Сендилом Мулайнатан из Чикагского университета:

  - Когда врачам показывали одинаковые медицинские карты, они гораздо реже рекомендовали пациентам афроамериканской расы проведение коронарографии (полезной процедуры).
  - При покупке подержанного автомобиля афроамериканцам предлагали первоначальные цены на 700 долларов выше, и они получали гораздо меньше скидок.
  - Ответы на объявления о сдаче квартир на Craigslist, отправленные с использованием афроамериканского имени, были менее многочисленными, чем ответы, отправленные с использованием белого имени.
  - Присяжные, состоящие только из белых людей, с вероятностью на 16 процентных пунктов выше выносили обвинительный приговор афроамериканскому подсудимому, чем белому, но когда в составе присяжных был хотя бы один афроамериканец, обвинительные приговоры выносились в одинаковой степени как афроамериканцам, так и белым.

Алгоритм COMPAS, широко используемый в США для определения приговоров и размера залога, является примером важного алгоритма, который, при тестировании организацией [ProPublica](https://www.propublica.org/article/machine-bias-risk-assessments-in-criminal-sentencing), продемонстрировал явную расовую предвзятость на практике (<<bail_algorithm>>).

Расовые предрассудки даже проявляются в компьютерном зрении, как показано на примере автоматически классифицированных фотографий, опубликованных пользователем Google Photos в Twitter (см. <<google_photos>>).

Система компании IBM, например, показывала 34,7% ошибок при распознавании темнокожих женщин, в то время как для светлокожих мужчин этот показатель составлял всего 0,3% – то есть более 100 раз больше ошибок! Некоторые люди неправильно интерпретировали результаты этих экспериментов, утверждая, что разница заключается просто в том, что компьютерам сложнее распознавать людей с более темной кожей. Однако, на самом деле, после негативной реакции общественности на этот результат, все компании, участвовавшие в разработке, значительно улучшили свои алгоритмы распознавания для людей с более темной кожей, так что через год они стали почти такими же эффективными, как и для людей со светлой кожей. Таким образом, это показало, что разработчики не использовали достаточное количество изображений людей с темной кожей в своих обучающих наборах данных или не тестировали свои продукты с использованием таких изображений.

Один из исследователей из MIT, Джой Буоламивини, предупредила: "Мы вступили в эпоху автоматизации, когда технологии становятся все более самоуверенными, но при этом недостаточно подготовленными. Если мы не создадим этичный и инклюзивный искусственный интеллект, мы рискуем потерять достижения в области гражданских прав и гендерного равенства, под предлогом нейтральности машин."

Частично проблема заключается в систематическом дисбалансе в составе популярных наборов данных, используемых для обучения моделей. В аннотации к статье ["No Classification Without Representation: Assessing Geodiversity Issues in Open Data Sets for the Developing World"](https://arxiv.org/abs/1711.08536), написанной Шреей Шанкаром и другими, говорится: "Мы анализируем два больших общедоступных набора изображений для оценки географического разнообразия и обнаруживаем, что эти наборы данных, по-видимому, демонстрируют заметное предубеждение в пользу американской и европейской культур. Кроме того, мы анализируем классификаторы, обученные на этих наборах данных, чтобы оценить влияние этих распределений данных, и обнаруживаем значительные различия в относительной производительности при обработке изображений из разных регионов." <<image_provenance>> показывает одну из графиков из статьи, показывающую географический состав двух наиболее важных наборов изображений, используемых для обучения моделей (на момент написания статьи и, по-прежнему, на момент написания этой книги).

#### Смещение при измерении.


В статье ["Does Machine Learning Automate Moral Hazard and Error"](https://scholar.harvard.edu/files/sendhil/files/aer.p20171084.pdf), опубликованной в журнале *American Economic Review*, Сендил Муллайнатан и Зиад Обермейер рассматривают модель, которая пытается ответить на вопрос: какие факторы, основываясь на исторических данных электронных медицинских карт (ЭМК), наиболее точно предсказывают риск инсульта? Вот основные факторы, выявленные моделью:

  - Перенесенный ранее инсульт
  - Сердечно-сосудистые заболевания
  - Случайные травмы
  - Доброкачественная опухоль молочной железы
  - Колоноскопия
  - Синусит

Однако, только первые два фактора действительно связаны с инсультом! Исходя из того, что мы изучили, вы, вероятно, можете догадаться, почему. Мы фактически не измеряли *инсульт*, который возникает, когда участок мозга лишается кислорода из-за нарушения кровоснабжения. Мы измеряли, кто имел симптомы, обращался к врачу, проходил необходимые обследования *и* получал диагноз "инсульт". Сам факт перенесенного инсульта – это не единственный фактор, связанный с этим списком. Он также связан с тем, что человек является тем, кто обычно обращается к врачу (что зависит от доступа к медицинской помощи, возможности оплачивать сопутствующие расходы, отсутствия дискриминации по расовому или гендерному признаку и т.д.)! Если вы склонны обращаться к врачу из-за *случайной травмы*, то, вероятно, вы обратитесь к врачу и в случае инсульта.

Это пример *смещения в измерениях*. Он возникает, когда наши модели ошибаются, потому что мы измеряем не то, что нужно, или измеряем это неправильным образом, или некорректно включаем эти измерения в модель.

#### Смещение, связанное с агрегированием данных.


*Смещение, связанное с агрегацией данных*, возникает, когда модели не агрегируют данные таким образом, чтобы учитывать все соответствующие факторы, или когда модель не включает необходимые параметры взаимодействия, нелинейные зависимости и т.д. Это может особенно часто происходить в медицинских условиях. Например, методы лечения диабета часто основаны на простых унивариантных статистических данных и исследованиях, в которых участвуют небольшие группы людей с различными характеристиками. Анализ результатов часто проводится таким образом, что не учитываются различия в этнической принадлежности или поле. Однако, как выясняется, у пациентов с диабетом наблюдаются [различные осложнения в зависимости от этнической принадлежности](https://www.ncbi.nlm.nih.gov/pubmed/24037313), а уровни гликированного гемоглобина (HbA1c), которые широко используются для диагностики и мониторинга диабета, [отличаются сложным образом в зависимости от этнической принадлежности и пола](https://www.ncbi.nlm.nih.gov/pubmed/22238408). Это может приводить к тому, что людям ставят неправильный диагноз или назначают неверное лечение, поскольку медицинские решения принимаются на основе модели, которая не учитывает эти важные переменные и взаимодействия.

#### Смещение, связанное с репрезентативностью выборки.


В аннотации к статье ["Bias in Bios: A Case Study of Semantic Representation Bias in a High-Stakes Setting"] (https://arxiv.org/abs/1901.09451), написанной Марией де-Артеагой и ее коллегами, отмечается, что существует гендерный дисбаланс в различных профессиях (например, женщины чаще работают медсестрами, а мужчины – пасторами), и говорится, что: "различия в показателях истинной положительности между разными полами коррелируют с существующим гендерным дисбалансом в профессиях, что может усугубить эти дисбалансы".

Другими словами, исследователи обнаружили, что модели, предсказывающие профессию, не только *отражают* фактический гендерный дисбаланс в изучаемой популяции, но и, по сути, *усиливают* его! Этот тип *смещения в представлении данных* довольно распространен, особенно для простых моделей. Когда существует четкая и легко заметная взаимосвязь, простая модель часто просто предполагает, что эта взаимосвязь всегда верна. Как показано в разделе <<representation_bias>> статьи, для профессий, в которых процент женщин был выше, модель, как правило, завышала распространенность этой профессии.

<img src="images/ethics/image12.png" id="representation_bias" caption="Ошибка модели при прогнозировании профессии, представленная в виде графика, где по оси абсцисс – процент женщин в данной профессии, а по оси ординат – величина ошибки." alt="График, показывающий, как прогнозы модели усиливают существующие предвзятости" width="500">

Например, в обучающем наборе данных 14,6% хирургов были женщинами, однако, согласно прогнозам модели, только 11,6% правильно определенных случаев относились к женщинам. Таким образом, модель усиливает предвзятость, существующую в обучающем наборе.

Теперь, когда мы убедились в существовании этих предвзятостей, что мы можем сделать, чтобы смягчить их?

### Борьба с различными видами предвзятости.

Различные типы предвзятости требуют различных подходов для их устранения. В то время как сбор более разнообразного набора данных может помочь в решении проблемы предвзятости, связанной с репрезентативностью, это не поможет в устранении исторических или систематических ошибок в измерениях. Все наборы данных содержат определенную степень предвзятости. Не существует абсолютно непредвзятого набора данных. Многие исследователи в этой области приходят к соглашению по ряду предложений, направленных на улучшение документации, касающейся решений, контекста и конкретных деталей о том, как и почему был создан определенный набор данных, в каких сценариях его уместно использовать и какие у него ограничения. Таким образом, пользователи конкретного набора данных не будут застигнуты врасплох его предвзятостями и ограничениями.

Мы часто слышим вопрос: "Люди предвзяты, так зачем вообще говорить об алгоритмической предвзятости?" Этот вопрос возникает настолько часто, что, должно быть, есть какая-то логика, которая кажется понятной тем, кто его задает, но нам это не кажется вполне логичным! Независимо от того, насколько это логично, важно понимать, что алгоритмы (особенно алгоритмы машинного обучения!) и люди – это разные вещи. Вот несколько моментов, касающихся алгоритмов машинного обучения:

  - _Машинное обучение может создавать петли обратной связи:_ Небольшие отклонения могут быстро увеличиваться в геометрической прогрессии из-за петель обратной связи.
  - _Машинное обучение может усиливать предвзятость:_ Человеческая предвзятость может приводить к увеличению предвзятости в системах машинного обучения.
  - _Алгоритмы и люди используются по-разному:_ Человеческие лица, принимающие решения, и алгоритмы, принимающие решения, не используются взаимозаменяемо на практике.
  - _Технологии – это сила:_ И вместе с этим приходит ответственность.

Как показал пример системы здравоохранения в Арканзасе, машинное обучение часто внедряется не потому, что оно приводит к лучшим результатам, а потому, что оно дешевле и эффективнее. Кэти О'Нил в своей книге "Оружие математического разрушения" (Crown) описала схему, при которой привилегированные люди обрабатываются людьми, а бедные – алгоритмами. Это лишь один из множества способов, которыми алгоритмы используются иначе, чем люди, принимающие решения. Другие примеры включают:

  - Люди чаще предполагают, что алгоритмы объективны или не содержат ошибок (даже если им предоставляется возможность вмешательства человека).
  - Алгоритмы чаще внедряются без возможности обжалования.
  - Алгоритмы часто используются в больших масштабах.
  - Алгоритмические системы – это дешево.

Даже в отсутствие предвзятости, алгоритмы (и особенно глубокое обучение, поскольку это очень эффективный и масштабируемый алгоритм) могут приводить к негативным социальным проблемам, например, когда они используются для *дезинформации*.

### Дезинформация


*Дезинформация* имеет долгую историю, насчитывающую сотни, а то и тысячи лет. Она не всегда направлена на то, чтобы заставить кого-то поверить в ложь, но часто используется для посева раздора и неопределенности, а также для того, чтобы люди отказались от поиска истины. Получение противоречивой информации может привести к тому, что люди начнут сомневаться в том, кому или чему можно доверять.

Некоторые считают, что дезинформация – это прежде всего ложная информация или *фейковые новости*, но на самом деле, дезинформация часто содержит элементы правды или полуправды, вырванные из контекста. Ладислав Биттман был сотрудником спецслужб в СССР, который впоследствии перешел на сторону США и написал несколько книг в 1970-х и 1980-х годах о роли дезинформации в советских пропагандистских операциях. В книге *КГБ и советская дезинформация* (Pergamon) он писал: "Большинство кампаний – это тщательно спланированная смесь фактов, полуправды, преувеличений и намеренных лжи".

В США эта проблема стала особенно актуальной в последние годы, когда ФБР раскрыло масштабную кампанию дезинформации, связанную с Россией, во время выборов 2016 года. Изучение дезинформации, использованной в этой кампании, очень познавательно. Например, ФБР установило, что российская кампания дезинформации часто организовывала две отдельные поддельные "народные" протесты, по одному для каждой стороны вопроса, и заставляла их проходить одновременно! Издание [*Houston Chronicle*](https://www.houstonchronicle.com/local/gray-matters/article/A-Houston-protest-organized-by-Russian-trolls-12625481.php) сообщило об одном из этих странных событий (<<texas>>).

С одной стороны улицы Трэвис, я увидел около 10 протестующих. С другой стороны, я насчитал около 50 контрпротестующих. Но я не смог найти организаторов акции. Никакого "Heart of Texas". Мне показалось это странным, и я упомянул об этом в статье: Как может группа не появиться на собственном мероприятии? Теперь я знаю, почему. Оказывается, организаторы акции находились в Санкт-Петербурге, в России, в то время. "Heart of Texas" – одна из интернет-групп, упомянутых в недавнем обвинительном заключении специального прокурора Роберта Мюллера в отношении россиян, пытавшихся вмешаться в президентские выборы в США.

Дезинформация часто включает в себя скоординированные кампании, основанные на поддельных действиях. Например, фиктивные аккаунты могут пытаться создать впечатление, что многие люди придерживаются определенной точки зрения. Хотя большинство из нас любят считать себя независимыми, на самом деле мы эволюционировали, чтобы подвергаться влиянию со стороны людей из нашей "внутренней" группы и, наоборот, противостоять тем, кто находится в нашей "внешней" группе. Онлайн-дискуссии могут влиять на наши взгляды или изменять рамки того, что мы считаем приемлемым. Люди – социальные существа, и как социальные существа мы подвержены огромному влиянию со стороны окружающих нас людей. Все чаще радикализация происходит в онлайн-среде; влияние исходит от людей, находящихся в виртуальном пространстве онлайн-форумов и социальных сетей.

Дезинформация, распространяемая с помощью автоматически генерируемого текста, является особенно важной проблемой, обусловленной значительно возросшими возможностями, предоставляемыми технологиями глубокого обучения. Мы подробно рассматриваем этот вопрос, когда переходим к созданию языковых моделей, в <<chapter_nlp>>.

Один из предложенных подходов заключается в разработке определенного вида цифровой подписи, внедрении ее в удобной форме и создании норм, согласно которым мы должны доверять только тому контенту, который был проверен. Директор института Allen Institute for AI, Орен Эцциони, представил такое предложение в статье под названием ["Как мы предотвратим подделки, созданные с использованием ИИ?"](https://hbr.org/2019/03/how-will-we-prevent-ai-based-forgery): "Искусственный интеллект может сделать высококачественные подделки дешевыми и автоматизированными, что может привести к потенциально катастрофическим последствиям для демократии, безопасности и общества. Угроза подделок, созданных с помощью ИИ, означает, что нам необходимо действовать, чтобы сделать цифровые подписи обязательными в качестве средства аутентификации цифрового контента."

Хотя мы не можем охватить все этические вопросы, которые поднимают технологии глубокого обучения и алгоритмы в целом, мы надеемся, что это краткое введение стало полезной отправной точкой, на которой вы сможете построить свои знания. Теперь мы перейдем к вопросам о том, как выявлять этические проблемы и что с ними делать.

## Выявление и решение этических проблем.

Ошибки случаются. Обнаружение этих ошибок и способы их устранения должны быть частью разработки любой системы, включающей машинное обучение (и многих других систем). Вопросы, возникающие в области этики данных, часто сложны и требуют междисциплинарного подхода, но крайне важно, чтобы мы работали над их решением.

Что же мы можем сделать? Это большая тема, но вот несколько шагов, которые помогут решить этические вопросы:

- Проанализируйте проект, над которым вы работаете.
- Внедрите в вашей компании процессы для выявления и устранения этических рисков.
- Поддерживайте правильную политику.
- Увеличивайте разнообразие.

Давайте рассмотрим каждый из этих шагов, начиная с анализа проекта, над которым вы работаете.

### Проанализируйте проект, над которым вы работаете.


При рассмотрении этических аспектов вашей работы легко упустить важные моменты. Один из способов, который может очень помочь, – это просто задавать правильные вопросы. Рэйчел Томас рекомендует учитывать следующие вопросы на протяжении всего процесса разработки проекта, связанного с данными:

- Стоит ли нам вообще это делать?
- Какие предвзятости присутствуют в данных?
- Можно ли проверить код и данные?
- Каковы показатели ошибок для разных подгрупп?
- Какова точность простого альтернативного решения, основанного на правилах?
- Какие процессы предусмотрены для рассмотрения жалоб или исправления ошибок?
- Насколько разнообразна команда, которая создала этот проект?

Эти вопросы могут помочь вам выявить нерешенные проблемы и возможные альтернативные решения, которые легче понять и контролировать. Помимо задавания правильных вопросов, важно также учитывать практики и процессы, которые необходимо внедрить.

На этом этапе важно учитывать, какие данные вы собираете и храните. Данные часто используются для целей, отличных от тех, для которых они были изначально собраны. Например, компания IBM начала продавать свою продукцию нацистской Германии задолго до Холокоста, в том числе помогала в проведении переписи населения Германии в 1933 году, организованной Адольфом Гитлером, которая позволила выявить гораздо больше евреев, чем было признано ранее в Германии. Аналогично, данные переписи населения США использовались для задержания американцев японского происхождения (которые были гражданами США) для содержания в лагерях во время Второй мировой войны. Важно осознавать, как собранные данные и изображения могут быть использованы в дальнейшем в деструктивных целях. Профессор Колумбийского университета [Тим Ву написал](https://www.nytimes.com/2019/04/10/opinion/sunday/privacy-capitalism.html), что "следует исходить из того, что любые персональные данные, которые Facebook или Android хранят, являются данными, которые правительства по всему миру попытаются получить, или которые воры попытаются украсть".

### Процессы для внедрения


Центр Маркулы опубликовал [Набор инструментов для этической практики в инженерии и дизайне](https://www.scu.edu/ethics-in-technology-practice/ethical-toolkit/), который включает в себя конкретные рекомендации для внедрения в вашей компании, в том числе регулярные проверки для выявления потенциальных этических рисков (по аналогии с тестированием на проникновение в области кибербезопасности), расширение круга лиц, чье мнение учитывается, для включения точек зрения различных заинтересованных сторон, и рассмотрение возможных негативных последствий (как злоумышленники могут использовать, украсть, исказить, взломать, уничтожить или превратить в оружие то, что вы создаете?).

Даже если у вас нет разнообразной команды, вы все равно можете попытаться учитывать мнения более широкой группы, задавая себе вопросы, предложенные центром Маркулы:

  - Чьи интересы, желания, навыки, опыт и ценности мы просто предположили, а не действительно учли?
  - Кто является всеми заинтересованными сторонами, на которых напрямую повлияет наш продукт? Как были защищены их интересы? Как мы можем узнать, каковы их истинные интересы — мы спросили их об этом?
  - Кто/какие группы и отдельные лица будут косвенно затронуты в значительной степени?
  - Кто может использовать этот продукт, но мы не ожидали, что он будет использоваться, или для целей, которые мы изначально не предполагали?

#### Этические аспекты.


Еще один полезный ресурс от Центра Маркулы – это его [Концептуальные рамки в области технологий и инженерной практики](https://www.scu.edu/ethics-in-technology-practice/ethical-lenses/). Этот документ рассматривает, как различные основополагающие этические подходы могут помочь выявить конкретные проблемы, и предлагает следующие подходы и ключевые вопросы:

  - Подход, основанный на правах: Какой вариант наилучшим образом учитывает права всех заинтересованных сторон?
  - Подход, основанный на справедливости: Какой вариант обеспечивает равное или пропорциональное отношение ко всем людям?
  - Утилитарный подход: Какой вариант принесет наибольшую пользу и причинит наименьший вред?
  - Подход, основанный на общественном благе: Какой вариант наилучшим образом служит интересам всего общества, а не только отдельных его членов?
  - Подход, основанный на добродетели: Какой вариант поможет мне действовать так, как я хочу, чтобы действовал?

Рекомендации Маркулы включают более глубокое изучение каждой из этих перспектив, в том числе рассмотрение проекта с точки зрения его *последствий*:

  - Кто непосредственно пострадает от этого проекта? Кто пострадает косвенно?
  - Вероятнее всего, приведут ли совокупные последствия этого проекта к большему благу, чем вреда, и какие именно виды блага и вреда?
  - Учитываем ли мы все соответствующие виды вреда/пользы (психологические, политические, экологические, моральные, когнитивные, эмоциональные, институциональные, культурные)?
  - Как этот проект может повлиять на будущие поколения?
  - Возлагаются ли риски причинения вреда от этого проекта непропорционально на наиболее уязвимые группы населения? Будут ли выгоды доставаться непропорционально обеспеченным слоям общества?
  - Достаточно ли мы рассмотрели возможность "двойного использования"?

Альтернативная точка зрения – это *деонтологическая* перспектива, которая сосредоточена на базовых понятиях *правильного* и *неправильного*:

  - Какие права других людей и какие обязанности перед другими мы должны соблюдать?
  - Как этот проект может повлиять на достоинство и автономию каждой заинтересованной стороны?
  - Какие соображения, касающиеся доверия и справедливости, имеют отношение к этому проекту/дизайну?
  - Связаны ли этот проект с какими-либо конфликтующими моральными обязательствами перед другими людьми или с конфликтующими правами заинтересованных сторон? Как мы можем расставить приоритеты в этих вопросах?

Один из лучших способов получения полных и продуманных ответов на такие вопросы – это обеспечение *разнообразия* среди людей, задающих эти вопросы.

### Сила разнообразия


В настоящее время, по данным [исследования компании Element AI](https://medium.com/element-ai-research-lab/estimating-the-gender-ratio-of-ai-researchers-around-the-world-81d2b8dbe9c3), менее 12% исследователей в области искусственного интеллекта — женщины. Статистика выглядит не лучше, когда речь идет о расовой принадлежности и возрасте. Когда все члены команды имеют схожий опыт, они, скорее всего, будут иметь схожие "слепые зоны" в отношении этических рисков. Издание *Harvard Business Review* (HBR) опубликовало ряд исследований, демонстрирующих многочисленные преимущества разнообразия в командах, в том числе:

- ["Как разнообразие может стимулировать инновации"](https://hbr.org/2013/12/how-diversity-can-drive-innovation)
- ["Команды решают проблемы быстрее, когда они более разнообразны с точки зрения когнитивных способностей"](https://hbr.org/2017/03/teams-solve-problems-faster-when-theyre-more-cognitively-diverse)
- ["Почему разнообразные команды умнее"](https://hbr.org/2016/11/why-diverse-teams-are-smarter), и
- ["Защитите свои исследования: что делает команду умнее? Больше женщин"](https://hbr.org/2011/06/defend-your-research-what-makes-a-team-smarter-more-women)

Разнообразие может способствовать более раннему выявлению проблем и рассмотрению более широкого спектра решений. Например, Трейси Чоу была одним из первых инженеров в компании Quora. Она [рассказывала о своем опыте](https://qz.com/1016900/tracy-chou-leading-silicon-valley-engineer-explains-why-every-tech-worker-needs-a-humanities-education/), описывая, как она выступала за добавление функции, которая позволяла бы блокировать троллей и других нежелательных пользователей. Чоу рассказывает: "Я была настроена работать над этой функцией, потому что лично чувствовала, что на сайте со мной обращались оскорбительно и грубо (пол — одна из вероятных причин)... Но если бы у меня не было этого личного опыта, возможно, команда Quora не придала бы первостепенного значения созданию кнопки блокировки на столь раннем этапе существования компании." Преследование часто вынуждает людей из маргинализированных групп покидать онлайн-платформы, поэтому эта функция была важна для поддержания здоровья сообщества Quora.

Важно понимать, что женщины покидают индустрию высоких технологий в два раза чаще, чем мужчины, согласно изданию [*Harvard Business Review*](https://www.researchgate.net/publication/268325574_By_RESEARCH_REPORT_The_Athena_Factor_Reversing_the_Brain_Drain_in_Science_Engineering_and_Technology) (41% женщин, работающих в сфере высоких технологий, уходят, по сравнению с 17% мужчин). Анализ более чем 200 книг, отчетов и статей показал, что причинами ухода являются "несправедливое отношение, более низкая заработная плата, меньшая вероятность продвижения по службе по сравнению с их коллегами-мужчинами и невозможность карьерного роста".

Исследования подтвердили ряд факторов, которые затрудняют продвижение женщин на рабочем месте. Женщины получают более расплывчатые отзывы и критику в отношении личности в оценках производительности, в то время как мужчины получают конкретные рекомендации, связанные с результатами деятельности (что более полезно). Женщины часто сталкиваются с тем, что их исключают из более творческих и инновационных ролей, и не получают задач, требующих высокого уровня ответственности и способствующих продвижению по службе. Одно исследование показало, что голоса мужчин воспринимаются как более убедительные, основанные на фактах и логичные, чем голоса женщин, даже при чтении одинаковых текстов.

Статистически доказано, что наставничество помогает мужчинам продвигаться, но не женщинам. Причина этого заключается в том, что, когда женщины получают наставничество, им дают советы о том, как им следует измениться и получить больше знаний о себе. Когда мужчины получают наставничество, это является публичным подтверждением их авторитета. Подумайте, что из этого более полезно для продвижения по службе?

Пока квалифицированные женщины продолжают уходить из сферы высоких технологий, простое обучение девочек программированию не решит проблемы с разнообразием, которые существуют в этой отрасли. Инициативы по продвижению разнообразия часто сосредотачиваются в основном на белых женщинах, даже несмотря на то, что женщины с другим цветом кожи сталкиваются с множеством дополнительных препятствий. В [интервью](https://worklifelaw.org/publications/Double-Jeopardy-Report_v6_full_web-sm.pdf) с 60 женщинами с другим цветом кожи, работающими в сфере STEM-исследований, 100% из них сообщали о случаях дискриминации.


Процесс найма особенно проблематичен в сфере технологий. Одно из исследований, указывающих на существующие недостатки, было проведено компанией Triplebyte, которая помогает размещать разработчиков программного обеспечения в компаниях, проводя стандартизированные технические собеседования в рамках этого процесса. У них есть очень интересный набор данных: результаты экзаменов более 300 разработчиков, сопоставленные с результатами их собеседований в различных компаниях. Главный вывод из [исследования Triplebyte](https://triplebyte.com/blog/who-y-combinator-companies-want) заключается в том, что "типы программистов, которых ищет каждая компания, часто мало связаны с тем, что компании действительно нужно или чем она занимается. Скорее, это отражает корпоративную культуру и опыт основателей."

Это создает определенные трудности для тех, кто стремится войти в сферу глубокого обучения, поскольку большинство групп, занимающихся глубоким обучением в компаниях, были созданы учеными. Эти группы, как правило, ищут людей, "похожих на них" – то есть, людей, которые могут решать сложные математические задачи и понимать сложный технический жаргон. Они не всегда умеют выявлять тех, кто действительно хорош в решении реальных задач с использованием глубокого обучения.

Это открывает большие возможности для компаний, которые готовы смотреть дальше формальных признаков и сосредоточиться на результатах!

### Честность, ответственность и прозрачность.

Профессиональное общество для ученых-компьютерщиков, ACM, проводит конференцию, посвященную этике данных, под названием "Конференция по вопросам справедливости, ответственности и прозрачности". Эта конференция ранее называлась *FAT*, но теперь использует менее вызывающее название *FAccT*. В компании Microsoft существует группа, занимающаяся вопросами "справедливости, ответственности, прозрачности и этики" (FATE). В этом разделе мы будем использовать термин "FAccT" для обозначения концепций *справедливости, ответственности и прозрачности*.

FAccT – это еще один инструмент, который может быть полезен при рассмотрении этических вопросов. Одним из полезных ресурсов является бесплатная онлайн-книга [*Справедливость и машинное обучение: ограничения и возможности*](https://fairmlbook.org/), написанная Солоном Барокасом, Морицем Хартом и Арвиндом Нараянаном, которая "предлагает взгляд на машинное обучение, в котором справедливость рассматривается как центральный вопрос, а не как второстепенный". Однако, в ней также предупреждается, что "она намеренно имеет узкий охват... Узкое понимание этики машинного обучения может быть соблазнительным для специалистов и компаний, поскольку это позволяет сосредоточиться на технических решениях, избегая при этом более глубоких вопросов о власти и ответственности. Мы предостерегаем от этой соблазнительной тенденции." Вместо того, чтобы давать обзор подхода FAccT к этике (что лучше сделать в таких книгах), мы сосредоточимся здесь на ограничениях такого узкого подхода.

Отличный способ оценить, является ли рассматриваемый этический подход всесторонним, – это попытаться привести пример, в котором этот подход и наши собственные этические представления дают разные результаты. Ос Кейс, Джеван Хатсон и Мерит Дербин исследовали это в своей статье ["Предложение о компостировании: анализ и улучшение алгоритмической системы, превращающей пожилых людей в высокопитательную массу"](https://arxiv.org/abs/1908.06166). В аннотации к статье говорится:


> Этические аспекты алгоритмических систем широко обсуждаются как в области взаимодействия человека и компьютера (HCI), так и в более широком сообществе, заинтересованном в проектировании, разработке и политике в области технологий. В этой статье мы исследуем применение одной из ведущих этических моделей – "Справедливость, Ответственность и Прозрачность" (Fairness, Accountability, and Transparency, FAT) – к предложенному алгоритму, который призван решать различные социальные проблемы, связанные с продовольственной безопасностью и старением населения. Используя различные стандартизированные методы аудита и оценки алгоритмов, мы значительно повысили соответствие алгоритма принципам модели FAT, что привело к созданию более этичной и полезной системы. Мы обсудим, как это может служить руководством для других исследователей или специалистов, стремящихся обеспечить более этичные результаты от использования алгоритмических систем в своей работе.

В данной работе весьма спортивное предложение ("превращение пожилых людей в питательную массу") и полученные результаты ("значительное повышение соответствия алгоритма принципам FAT, что приводит к более этичной и полезной системе") противоречат друг другу... мягко говоря!

В философии, особенно в этике, это один из самых эффективных методов: сначала предлагается процесс, определение, набор вопросов и т.д., предназначенные для решения определенной проблемы. Затем следует придумать пример, в котором кажущееся решение приводит к предложению, которое никто не сочтет приемлемым. Это может привести к дальнейшей доработке решения.

До сих пор мы сосредоточивались на том, что вы и ваша организация можете сделать. Но иногда индивидуальных или организационных действий недостаточно. Иногда правительствам также необходимо учитывать политические последствия.

## Роль политики


Мы часто общаемся с людьми, которые надеются, что технические или дизайнерские решения станут полноценным решением проблем, о которых мы говорим. Например, технический подход к устранению предвзятости в данных или рекомендации по дизайну, направленные на то, чтобы технологии были менее вызывающими привыкание. Хотя такие меры могут быть полезными, их будет недостаточно для решения основных проблем, которые привели нас к нынешнему положению дел. Например, пока создание вызывающих привыкание технологий остается чрезвычайно прибыльным, компании будут продолжать это делать, независимо от того, приводит ли это к распространению теорий заговора и загрязнению нашей информационной среды. Даже если отдельные дизайнеры будут пытаться вносить небольшие изменения в дизайн продуктов, существенных изменений мы не увидим, пока не изменятся основные мотивы, определяющие прибыльность.

### Эффективность регулирования.

Чтобы понять, какие факторы могут побудить компании к конкретным действиям, рассмотрим два примера того, как вела себя компания Facebook. В 2018 году расследование ООН показало, что Facebook сыграла "решающую роль" в продолжающемся геноциде народа рохинджа, этнического меньшинства в Мьянме, которого Генеральный секретарь ООН Антониу Гутерриш назвал "одним из, если не самым, дискриминируемых народов в мире". Местные активисты еще в 2013 году предупреждали руководство Facebook о том, что их платформа используется для распространения ненавистнических высказываний и подстрекательства к насилию. В 2015 году им сообщили, что Facebook может сыграть ту же роль в Мьянме, что и радиовещание во время геноцида в Руанде (когда были убиты миллионы людей). Однако к концу 2015 года в Facebook работало всего четыре сотрудника, владеющих бирманским языком. Как сказал один человек, близкий к этой проблеме, "Это не просто анализ после произошедшего. Масштаб этой проблемы был значительным, и это было уже очевидно". Цукерберг пообещал во время слушаний в Конгрессе нанять "десятки" сотрудников для решения проблемы геноцида в Мьянме (в 2018 году, спустя годы после начала геноцида, включая уничтожение огнем не менее 288 деревень в северной части штата Ракхайн после августа 2017 года).

Это резко контрастирует с тем, как Facebook быстро [наняла 1200 человек в Германии](http://thehill.com/policy/technology/361722-facebook-opens-second-german-office-to-comply-with-hate-speech-law), чтобы избежать дорогостоящих штрафов (до 50 миллионов евро) в соответствии с новым немецким законом, направленным на борьбу с разжиганием ненависти. Очевидно, что в данном случае Facebook больше реагировала на угрозу финансовых санкций, чем на систематическое уничтожение этнического меньшинства.

В [статье о проблемах конфиденциальности](https://idlewords.com/2019/06/the_new_wilderness.htm) Мачей Цеглоский проводит параллели с экологическим движением:

> : Этот регуляторный проект оказался настолько успешным в развитых странах, что мы рискуем забыть, какой была жизнь до него. Удушливый смог, подобный тому, что сегодня убивает тысячи людей в Джакарте и Дели, [когда-то был] характерной чертой Лондона. Река Куяхога в штате Огайо [регулярно] загоралась. В особенно ужасном примере непредвиденных последствий, добавление тетраэтилсвинца в бензин [привело к] росту уровня насильственной преступности во всем мире в течение пятидесяти лет. Ни одно из этих негативных последствий нельзя было устранить, просто призывая людей делать выбор, основываясь на своих финансовых интересах, или тщательно изучать экологическую политику каждой компании, с которой они сотрудничают, или прекратить использование определенных технологий. Для решения этих проблем требовалось скоординированное, а иногда и очень сложное регулирование, выходящее за рамки юрисдикций. В некоторых случаях, как, например, в случае [запрета на использование хладагентов, разрушающих озоновый слой](https://en.wikipedia.org/wiki/Montreal_Protocol), это регулирование требовало всемирного консенсуса. Мы находимся на этапе, когда нам необходимо аналогичное изменение перспективы в нашем законодательстве о конфиденциальности.


### Права и политика.

Чистый воздух и чистая питьевая вода – это общественные блага, которые практически невозможно защитить с помощью индивидуальных рыночных решений, а скорее требуют скоординированных регуляторных мер. Аналогично, многие негативные последствия, возникающие в результате непреднамеренных последствий неправильного использования технологий, связаны с общественными благами, такими как загрязненная информационная среда или ухудшение условий приватности. Слишком часто приватность рассматривается как индивидуальное право, однако широкомасштабное наблюдение оказывает влияние на общество (и это было бы так даже в том случае, если бы некоторым отдельным лицам было бы возможно отказаться от него).

Многие проблемы, которые мы наблюдаем в сфере технологий, на самом деле являются вопросами прав человека, например, когда предвзятый алгоритм рекомендует более длительные сроки тюремного заключения для чернокожих обвиняемых, когда определенные объявления о работе показываются только молодым людям, или когда полиция использует распознавание лиц для идентификации протестующих. Соответствующей площадкой для решения вопросов прав человека, как правило, является законодательство.

Нам необходимы как регуляторные, так и законодательные изменения, *и* этичное поведение отдельных лиц. Изменение индивидуального поведения не может решить проблемы, связанные с неправильными мотивами получения прибыли, внешними эффектами (когда корпорации получают большую прибыль, перекладывая свои затраты и негативные последствия на более широкое общество) или системными сбоями. Однако закон никогда не сможет охватить все возможные случаи, и важно, чтобы отдельные разработчики программного обеспечения и специалисты по данным были подготовлены к принятию этических решений на практике.

### Автомобили: Исторический Прецедент.


Проблемы, с которыми мы сталкиваемся, сложны, и не существует простых решений. Это может быть демотивирующим, но мы находим надежду, рассматривая другие масштабные вызовы, которые люди успешно преодолевали на протяжении истории. Одним из примеров является движение за повышение безопасности автомобилей, которое рассматривается в качестве кейс-стади в работе Тимнит Гебру и других, опубликованной под названием ["Datasheets for Datasets"](https://arxiv.org/abs/1803.09010), а также в подкасте [99% Invisible](https://99percentinvisible.org/episode/nut-behind-wheel/). В ранних автомобилях не было ремней безопасности, металлические ручки на приборной панели, которые могли застрять в черепе при аварии, обычные стеклянные окна, которые разбивались опасным образом, и неподвижные рулевые колонки, которые протыкали водителей. Однако компании-производители автомобилей были крайне невосприимчивы даже к обсуждению идеи безопасности как проблемы, которую они могли бы решить, и широко распространено было мнение, что автомобили просто такие, какие они есть, и что проблемы возникают из-за людей, которые ими пользуются.

Потребовались десятилетия работы активистов и защитников прав потребителей, чтобы изменить общественное мнение и добиться того, чтобы начали рассматривать возможность того, что компании-производители автомобилей несут определенную ответственность, которую необходимо решать посредством регулирования. Когда была изобретена складная рулевая колонка, она не была внедрена в течение нескольких лет, поскольку не было никакой финансовой выгоды от этого. Крупнейшая автомобильная компания General Motors наняла частных детективов, чтобы попытаться найти компромат на защитника прав потребителей Ральфа Наддера. Требования по использованию ремней безопасности, манекенов для краш-тестов и складных рулевых колонок стали важными победами. Только в 2011 году компаниям-производителям автомобилей было предписано начинать использовать манекены для краш-тестов, которые представляли бы среднюю женщину, а не только средние мужские тела; до этого женщины с 40% большей вероятностью получали травмы в автомобильной аварии той же силы, чем мужчины. Это яркий пример того, как предвзятость, политика и технологии могут иметь важные последствия.

## Заключение


Имея опыт работы с двоичной логикой, отсутствие четких ответов в области этики может сначала казаться разочаровывающим. Однако, последствия нашей работы, влияющие на мир, включая непредвиденные последствия и использование наших разработок в злонамеренных целях, – это одни из самых важных вопросов, которые мы должны (и должны!) рассматривать. Даже если нет простых ответов, существуют определенные ловушки, которых следует избегать, и практики, которым стоит следовать, чтобы двигаться к более этичному поведению.

Многие люди (включая нас!) ищут более убедительные и надежные ответы о том, как смягчить негативное воздействие технологий. Однако, учитывая сложный, масштабный и междисциплинарный характер проблем, с которыми мы сталкиваемся, не существует простых решений. Джулия Энгвин, бывший старший журналист издания ProPublica, специализирующаяся на вопросах алгоритмической предвзятости и слежки (и одна из исследователей алгоритма COMPAS, используемого для оценки вероятности рецидива, что и стало отправной точкой для развития области FAccT), говорила в [интервью 2019 года](https://www.fastcompany.com/90337954/who-cares-about-liberty-julia-angwin-and-trevor-paglen-on-privacy-surveillance-and-the-mess-were-in):

> : Я твердо убеждена, что для решения проблемы необходимо сначала ее диагностировать, и что мы все еще находимся на стадии диагностики. Если вспомнить начало века и индустриализацию, у нас было, наверное, 30 лет эксплуатации детского труда, бесконечной рабочей нагрузки и ужасных условий труда. Для того чтобы понять суть проблемы и найти решение, потребовались усилия журналистов, расследований и адвокации, а затем – активизм для изменения законов. Мне кажется, мы переживаем вторую индустриализацию данных и информации... Я вижу свою роль в том, чтобы максимально четко и точно выявлять негативные стороны и диагностировать их, чтобы их можно было решить. Это сложная работа, и гораздо больше людей должны этим заниматься.

То, что Энгвин считает, что мы в основном все еще находимся на стадии диагностики, вселяет уверенность: если ваше понимание этих проблем кажется неполным, это нормально и естественно. У еще ни у кого нет "лекарства", хотя крайне важно, чтобы мы продолжали работать над более глубоким пониманием и решением проблем, с которыми мы сталкиваемся.

Один из наших рецензентов этой книги, Фред Монро, раньше работал в сфере торговли ценными бумагами. Он сказал нам, прочитав эту главу, что многие из обсуждаемых здесь проблем (например, значительное отличие распределения данных от данных, на которых была обучена модель, влияние обратной связи на модель после ее внедрения и масштабирования и т.д.) также являются ключевыми для создания прибыльных торговых моделей. Меры, которые необходимо предпринять для учета социальных последствий, во многом совпадают с мерами, которые необходимо предпринять для учета организационных, рыночных и клиентских последствий. Таким образом, тщательное обдумывание этических вопросов также может помочь вам более внимательно подойти к тому, как сделать ваш продукт на основе данных успешным в целом!
